## Abstract
I always loved watching NBA games and wondered whether resting stars during the regular season actually gave teams a meaningful edge come playoff time, or whether it was just something good teams did because they could afford to. When I dug deeper, the data told a pretty clear story: once I controlled for how good a team actually was, rest had essentially no association with winning. The apparent correlation was a data artifact from inconsistent reporting in the early seasons, not a real basketball effect. As a secondary question I also looked at whether the NBA's 2023 Player Participation Policy changed how teams handled star rest, and while the directional evidence suggested it did, the statistical evidence was borderline rather than conclusive given how short the post-policy window was.

The bigger takeaway was that load management is largely a luxury of strong organizations, whhere good teams have stars worth resting, and good teams win games, but the two move together because of roster quality rather than because rest itself drives outcomes. The policy meanwhile appeared to have nudged star rest rates downward without meaningfully disrupting that underlying relationship, which is roughly what it was designed to do.

## Research Question

Among NBA teams from the 2017–18 through 2024–25 regular seasons, is the rate of strategic rest games among a team's top-3 minutes leaders associated with regular season win percentage and playoff seeding, after controlling for team strength (preseason over/under win total), roster age, and back-to-back game frequency? 

As a secondary analysis, we explore whether the NBA's 2023 Player Participation Policy - which fines teams for resting "star" players (those named to All-Star or All-NBA teams in the prior three seasons) in nationally televised games,appears to have changed rest patterns for policy-eligible stars relative to non-star high-minutes players.


## Background and Prior Work

The practice of "load management," strategically resting healthy NBA players during the regular season to preserve them for the playoffs, has become one of the most contentious topics in modern professional basketball. Once an obscure concept used by a small number of teams, load management entered mainstream attention when Kawhi Leonard sat out 22 of 82 regular season games during the Toronto Raptors' 2018-19 championship season, an approach widely credited with helping him win Finals MVP.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Since then, the practice has spread league-wide, generating concerns from broadcasters, fans, and the league office about competitive integrity and the value of regular-season games. The fundamental question for our project is whether this strategy is associated with team success, and whether the league's recent attempt to curb the practice has changed team behavior.

The empirical literature on regular-season rest and player outcomes is small but growing, and notably its findings have not strongly supported the assumed benefits of load management. Belk and colleagues conducted one of the earliest systematic analyses, examining 811 NBA players over the 2005-06 through 2014-15 seasons using DNP-CD ("did not play, coach's decision") records from Basketball Reference and Pro Sports Transactions.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Comparing players who missed fewer than 5 regular-season games due to rest against those who missed 5 to 9 games, they found no significant differences in playoff performance metrics or injury risk. The authors concluded that, contrary to widespread belief, regular-season rest did not improve playoff outcomes for the rested players themselves.

A more recent study by Pradhan and Miller extended this work using data from the 2016-17 through 2020-21 seasons, with rest decisions sourced from Spotrac.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Using Player Efficiency Rating and Win Shares as outcome measures, they similarly found insufficient evidence that strategic rest during the regular season translated into better post-season performance. A nested case-control comparison of 184 players (92 cases of rested players matched with 92 controls) showed no meaningful differences in performance changes from regular to post-season. Importantly, both studies focused on individual player performance rather than team-level outcomes such as win percentage or playoff seeding, and neither examined how the practice's prevalence has shifted over time. Our project addresses these gaps directly.

In response to growing concerns about load management's effect on regular-season competitiveness and broadcast value, the NBA Board of Governors in September 2023 approved the Player Participation Policy, which took effect at the start of the 2023-24 season.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) The policy formally defines a "star player" as any player named to an All-Star or All-NBA team in any of the prior three seasons, and prohibits teams from resting these players in nationally televised games or in-season tournament games without approved justification. Violations carry escalating fines beginning at USD 100,000 for a first offense and rising to USD 1.25 million for a third. Bobby Marks of ESPN provided detailed coverage of the policy's mechanics, including the specific rest restrictions, exception categories for older or high-mileage stars, and the league's investigative authority over suspect injury designations.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) This policy creates a natural setting for examining whether legal restrictions on rest decisions actually change team behavior, a question that has not yet been studied empirically given that the policy is recent.

Our project builds on the prior literature in three specific ways. First, we shift the unit of analysis from individual players to teams, asking whether a team's overall rest patterns are associated with regular-season win percentage and playoff seeding rather than individual player performance. Second, we extend the time window through the 2024-25 season, capturing both the pre- and post-policy eras. Third, by leveraging the policy's binary nature (it applies only to players meeting specific All-Star or All-NBA criteria), we can compare rest patterns for policy-eligible stars against those of non-star high-minutes players over the same period. This comparison group acts as a baseline to help separate the policy's effect from any general league-wide trends in rest behavior.

## Hypothesis


We predict that higher rates of strategic rest among a teams top-3 minutes leaders depict a weak positive association with regular-season win percentage, and that teams with resting stars will likely achieve higher playoff seeding after controlling for team strength, roster age, and back-to-back frequency. We base this on the expectation that load management is largely a luxury of strong organizations, meaning rest rate may partly reflect roster quality rather than cause wins or losses. We additionally predict that the 2023 Player Participation Policy will have reduced league-wide rest rates without meaningfully altering this underlying relationship.

### Data Overview

**Dataset #1 — NBA Team Statistics and Game Logs (2016-17 to 2024-25)**

| Field | Detail |
|---|---|
| Link | https://www.basketball-reference.com/leagues/ |
| Observations | 9,840 games across 8 seasons (~1,230 per season) |
| Variables | 9 |

**Key variables:**
- `away team` / `home team` — the two teams playing
- `away score` / `home score` — final points scored by each team
- `date` — when the game was played
- `attendance` — number of fans at the game
- `arena` — where the game was played

**Shortcomings:** Does not include player-level DNP records needed to identify strategic rest. Post-2023 data may undercount rest due to teams mislabeling rest as injury to avoid NBA fines.

---

**Dataset #2 — NBA Player Injury and Rest Transaction Log (2016-2025)**

| Field | Detail |
|---|---|
| Link | https://www.kaggle.com/datasets/jacquesoberweis/2016-2025-nba-injury-data |
| Observations | ~18,000+ injury/transaction records |
| Variables | 5 |

**Key variables:**
- `Date` — date of the transaction; lets us match absences to specific games
- `Team` — which team the player belongs to
- `Relinquished` — player being placed on the injury list (who sat out)
- `Acquired` — player being activated back from the injury list
- `Notes` — reason for the absence *(e.g. "placed on IL for rest", "sore knee")* — **most important column** for identifying strategic rest vs real injury

**Shortcomings:** Notes are self-reported by teams, so strategic rest can be disguised as minor injury — especially post-2023 when teams face fines. Some entries have vague or missing notes. The data records when a player was placed on/activated from the list, not how many games they actually missed.

---

**Dataset #3 — NBA Player Per-Game Statistics (2017-18 to 2024-25, Basketball Reference)**

| Field | Detail |
|---|---|
| Link | https://www.basketball-reference.com/leagues/NBA_2024_per_game.html (year changes per season) |
| Observations | 5,697 raw rows → 5,079 clean player-team-season rows after removing header/aggregate rows |
| Variables | 32 raw + 2 derived (total_mp, season) |

**Key variables:**
- `Player` — player name; join key to Dataset #2
- `Team` — BBRef abbreviation (e.g. LAL, BOS, GSW)
- `G` — games played for that team that season
- `MP` — minutes per game average
- `total_mp` *(derived)* — MP × G; used to rank and identify each team's top-3 minutes leaders
- `Awards` — concatenated award codes (AS = All-Star, NBA1/NBA2/NBA3 = All-NBA); used to flag policy-eligible stars

**Shortcomings:** MP is a per-game average so total minutes must be derived. Traded players appear in multiple rows — aggregate rows must be dropped before ranking. Awards column has no separators between codes, requiring careful string parsing.

---

**Combining all three datasets:**

The final merge uses **team abbreviation + season** as join keys across all three datasets. Datasets #1 and #3 already use BBRef abbreviations natively. Dataset #2 gets a `team_abbr` column via a mascot-to-abbreviation mapping so all three share the same key format.

The merged table has **one row per team-season (240 rows total)** with:
- `rest_rate_top3` — from Datasets #2 and #3 
- `win_pct`, `srs`, `roster_age`, `pace` — from Dataset #1 
- `post_policy` — derived flag for seasons after the 2023 NBA Player Participation Policy

```python
# =====================================================================
# 3A-3G: Load, Tidy, Audit, and Clean — NBA Per-Game Player Stats (Dataset #3)
# =====================================================================
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- 3A: LOAD ALL 8 SEASON CSVs ---------------------------------------
RAW_DIR = 'data/00-raw'

all_dfs = []
csv_files = sorted(glob.glob(os.path.join(RAW_DIR, 'nba_*_per_game.csv')))

print("=== 3A: RAW LOAD ===")
print(f"Found {len(csv_files)} season files:")

for fpath in csv_files:
    fname = os.path.basename(fpath)
    parts = fname.split('_')
    season_year = int(parts[2]) + 2000
    tmp = pd.read_csv(fpath)
    tmp['season'] = season_year
    all_dfs.append(tmp)
    print(f"  {fname}  ->  season={season_year}, shape={tmp.shape}")

df = pd.concat(all_dfs, ignore_index=True)
print(f"\nCombined raw shape: {df.shape}")

# --- 3B: TIDY ---------------------------------------------------------
print("\n=== 3B: TIDY ===")

for col in ['Player', 'Team', 'Pos']:
    df[col] = df[col].astype(str).str.strip().replace({'nan': np.nan, '': np.nan})

df['G']  = pd.to_numeric(df['G'],  errors='coerce')
df['MP'] = pd.to_numeric(df['MP'], errors='coerce')

# Drop repeated BBRef header rows
n_before = len(df)
df = df[pd.to_numeric(df['Rk'], errors='coerce').notna()].reset_index(drop=True)
print(f"Dropped {n_before - len(df)} repeated header rows. Rows now: {len(df)}")

# Drop aggregate rows for traded players — keep per-team rows only
# Each player's minutes are attributed to the specific team they played for
n_before = len(df)
df = df[~df['Team'].isin(['TOT', '2TM', '3TM', '4TM'])].reset_index(drop=True)
print(f"Dropped {n_before - len(df)} aggregate rows for traded players. Rows now: {len(df)}")

# Compute total minutes: MP (per game average) x G (games played for that team)
df['total_mp'] = df['MP'] * df['G']

print("Tidy check: one row = one player-team-season")
print(df[['Player', 'Team', 'season', 'G', 'MP', 'total_mp']].head(8).to_string(index=False))

# --- 3C: SIZE AUDIT ---------------------------------------------------
print("\n=== 3C: DATASET SIZE ===")
print(f"Total player-team-season rows : {len(df):,}")
print(f"Unique players                : {df['Player'].nunique():,}")
print(f"Unique teams                  : {df['Team'].nunique()}")
print(f"Seasons covered               : {sorted(df['season'].unique())}")

multi = df.groupby(['Player', 'season'])['Team'].nunique()
print(f"\nPlayer-seasons split across 2+ teams (traded): {(multi > 1).sum()}")

print("\nRows per season:")
print(df['season'].value_counts().sort_index())

# --- 3D: MISSINGNESS AUDIT --------------------------------------------
print("\n=== 3D: MISSING DATA AUDIT ===")
print("Missing values per column:")
print(df[['Player', 'Team', 'season', 'G', 'MP', 'total_mp']].isna().sum())

os.makedirs('data/01-interim', exist_ok=True)
plt.figure(figsize=(12, 5))
sns.heatmap(df[['Player', 'Team', 'G', 'MP', 'total_mp', 'season']].isna(),
            cbar=False, yticklabels=False, cmap='viridis')
plt.title('Missing data -- NBA per-game stats (2017-18 through 2024-25)')
plt.tight_layout()
plt.savefig('data/01-interim/dataset3_missingness.png', dpi=150, bbox_inches='tight')
plt.show()

n_before = len(df)
df = df.dropna(subset=['Player', 'MP']).reset_index(drop=True)
print(f"Dropped {n_before - len(df)} rows missing Player or MP. Rows now: {len(df)}")

# --- 3E: OUTLIERS + SUSPICIOUS ENTRIES --------------------------------
print("\n=== 3E: OUTLIERS + SUSPICIOUS ENTRIES ===")

outlier_mp = df[df['MP'] > 48]
print(f"Rows with MP > 48 (impossible in regulation): {len(outlier_mp)}")
if len(outlier_mp) > 0:
    print(outlier_mp[['Player', 'Team', 'season', 'G', 'MP']].to_string(index=False))

low_games = df[df['G'] < 20]
print(f"\nPlayer-team stints with fewer than 20 games: {len(low_games)}")
print("  (excluded from top-3 ranking -- not genuine rotation players for that team)")

print("\nMedian MP per game by season:")
print(df.groupby('season')['MP'].median().round(1))

# --- 3F: TOP-3 MINUTES LEADERS LOOKUP TABLE --------------------------
print("\n=== 3F: TOP-3 MINUTES LEADERS LOOKUP TABLE ===")

# Require >= 20 games for that specific team stint to qualify
df_qualified = df[df['G'] >= 20].copy()
print(f"Player-team stints with >= 20 games (qualified): {len(df_qualified):,}")

df_qualified['rank_mp'] = (
    df_qualified
    .groupby(['season', 'Team'])['total_mp']
    .rank(method='first', ascending=False)
)

top3 = (
    df_qualified[df_qualified['rank_mp'] <= 3]
    [['season', 'Team', 'Player', 'G', 'MP', 'total_mp', 'rank_mp']]
    .sort_values(['season', 'Team', 'rank_mp'])
    .reset_index(drop=True)
)

expected = 30 * 8 * 3
print(f"\nTop-3 table shape: {top3.shape}  (expected {expected} rows)")
print(f"All 30 teams present every season: {(top3.groupby('season')['Team'].nunique() == 30).all()}")

print("\nSample -- 2024 Golden State Warriors:")
print(top3[(top3['Team'] == 'GSW') & (top3['season'] == 2024)].to_string(index=False))

print("\nSample -- 2020 LA Lakers:")
print(top3[(top3['Team'] == 'LAL') & (top3['season'] == 2020)].to_string(index=False))

# --- 3G: SAVE ---------------------------------------------------------
print("\n=== 3G: SAVE ===")
os.makedirs('data/02-processed', exist_ok=True)

full_path = 'data/02-processed/nba_player_stats_clean.csv'
df.to_csv(full_path, index=False)
print(f"Full cleaned stats table  ->  {full_path}  {df.shape}")

top3_path = 'data/02-processed/nba_top3_leaders_by_team_season.csv'
top3.to_csv(top3_path, index=False)
print(f"Top-3 lookup table        ->  {top3_path}  {top3.shape}")

print("\n=== CLEANING JUSTIFICATION ===")
print(
    "1. Dropped repeated BBRef header rows (Rk = 'Rk') -- formatting artifacts\n"
    "   from the HTML export, not real player rows.\n"
    "2. Dropped aggregate rows (TOT, 2TM, 3TM, 4TM) and kept per-team rows for\n"
    "   traded players. Each player's minutes are attributed to the specific team\n"
    "   they played for, so a traded player can qualify for the top-3 of each team\n"
    "   they contributed to based on their games played for that team specifically.\n"
    "3. Dropped rows missing Player name or MP -- no useful data for ranking.\n"
    "4. Top-3 ranking requires >= 20 games for that specific team stint -- excludes\n"
    "   short-term signings and 10-day contracts who are not genuine rotation players.\n"
    "5. Ranked by total_mp (MP x G), not per-game MP alone -- correctly weights\n"
    "   players who played many games over those with few high-minute appearances."
)

print(f"\nFinal full table : {df.shape}")
print(f"Final top-3 table: {top3.shape}")
```

## Question 2a:
Dataset three is a player-level per-game statistics table scraped from Basketball 
Reference, with one CSV file per NBA regular season from 2017-18 through 2024-25. 
Each row in the cleaned dataset represents a single player's statistics for one 
team during one season. After combining all 8 files we start with 5,697 raw rows. 
We drop 8 repeated header rows (a Basketball Reference HTML formatting artifact) 
and 610 aggregate rows created for players who were traded mid-season, leaving 
5,079 clean player-team-season rows covering 1,291 unique players across all 30 
NBA teams. The purpose of this dataset in our project is specifically to identify, 
for every team-season, the three players who led that team in total minutes played. 
Those three players are the subjects of our load management analysis. The key 
variables we use are:

- **Player:** Player name, used as the join key to Dataset #2 after light name 
standardization for special characters and suffixes.

- **Team:** The team abbreviation for that specific row (e.g. LAL, BOS, GSW). 
For traded players we keep one row per team they played for, so their minutes 
are attributed to each specific team rather than combined into a season total.

- **G (Games Played):** Number of regular-season games the player appeared in 
for that specific team. A player traded mid-season will have a separate G value 
for each team they played for. We require a minimum of 20 games for a player to 
qualify for a team's top-3 ranking, since anything below that is a short-term 
stint rather than a genuine rotation role.

- **MP (Minutes Per Game):** Average minutes played per game for that team. 
This is a per-game average, not a season total. To get total minutes we multiply 
MP by G.

- **total_mp (derived):** Total minutes played for that team, computed as MP 
times G. This is the variable we rank players by within each team-season to 
identify the top-3 minutes leaders. We use total minutes rather than per-game 
average because a player who averaged 38 minutes in 10 games should not outrank 
a player who played all 70 games at 30 minutes per game.

- **Awards:** A concatenated string of any honors the player received that season, 
for example AS for All-Star and NBA1, NBA2, NBA3 for All-NBA First, Second, and 
Third Team. We parse this column to flag players who are eligible under the 2023 
NBA Player Participation Policy, which defines a star as any player who received 
one of these honors in any of the three prior seasons.

The final top-3 lookup table contains exactly 720 rows, 3 players for each of 
the 30 teams across each of the 8 seasons, with all 30 teams represented in 
every season.

## Question 2b:
This dataset has two real concerns and one structural decision worth documenting.

The first concern is how we handle players traded mid-season. When a player is 
traded, Basketball Reference creates multiple rows: one per team they played for, 
plus aggregate rows labeled TOT, 2TM, or 3TM that combine their full-season 
stats. We drop all aggregate rows and keep only the per-team rows. This means 
each player's minutes are attributed to the specific team they played for, and a 
traded player can qualify for the top-3 of each team they played on, provided 
they meet the 20-game minimum for that specific team stint. Across our 8 seasons 
there are 610 such player-seasons split across 2 or more teams, which is why the 
raw data has 5,689 rows but only 5,079 after removing aggregate rows. The 
trade-off of this approach is that a player traded mid-season with 18 games on 
each team would be excluded from both teams' top-3 rankings since neither stint 
meets the 20-game threshold, even though they were a meaningful player for parts 
of both seasons. We accept this as a reasonable limitation since a player with 
fewer than 20 games on a team is genuinely not one of that team's primary 
rotation players for that season.

The second concern is the 20-game minimum threshold itself. We chose 20 games 
because it filters out short-term signings, 10-day contracts, and brief injury 
returns who are not genuine high-usage starters. The output shows 1,479 
player-team stints below this threshold, meaning roughly 29% of all per-team 
rows are excluded from the ranking. This is expected and correct behavior since 
many NBA roster spots are filled by players who appear in only a handful of games.

Finally, one non-issue worth naming: the dataset shows zero missing values across 
all key columns (Player, Team, G, MP, total_mp, season) after cleaning, confirmed 
by both the missingness table and the completely uniform heatmap. The dataset is 
spotless for the variables we care about.

### Dimensionality Reduction via PCA

PCA on seven standardized team statistics (SRS, offensive/defensive ratings, pace, roster age, points scored/allowed per game) yields two principal components explaining 77.4% of variation. PC1 (40.5%) represents overall team quality, with offensive rating, defensive rating, and SRS loading heavily. PC2 (36.9%) captures secondary patterns, likely pace-related.

PC1 correlates extremely strongly with win percentage (r = 0.94) and playoff seeding (r = -0.91), while PC2 shows weaker relationships with outcomes.

```python
# Plot 6: PCA combines multiple stats into two main dimensions.
pcaColumns = [
    "srs", "offensive_rating", "defensive_rating",
    "pace", "roster_age", "points_scored_per_game",
    "points_allowed_per_game"
]

pcaData = df[pcaColumns].dropna()

scaler = StandardScaler()
scaledData = scaler.fit_transform(pcaData)

pcaModel = PCA(n_components=2)
pcaScores = pcaModel.fit_transform(scaledData)

pcaTable = df.loc[pcaData.index, ["team", "season", "win_pct", "playoff_seed"]].copy()
pcaTable["PC1"] = pcaScores[:, 0]
pcaTable["PC2"] = pcaScores[:, 1]

loadingTable = pd.DataFrame(
    pcaModel.components_.T,
    columns=["PC1", "PC2"],
    index=pcaColumns
)

plt.figure(figsize=(9, 7))
sns.scatterplot(data=pcaTable, x="PC1", y="PC2", hue="win_pct", palette="viridis", s=70)

for statName in loadingTable.index:
    plt.arrow(0, 0, loadingTable.loc[statName, "PC1"] * 3, loadingTable.loc[statName, "PC2"] * 3,
              color="black", alpha=0.7, head_width=0.06)
    plt.text(loadingTable.loc[statName, "PC1"] * 3.2, loadingTable.loc[statName, "PC2"] * 3.2,
             statName, fontsize=9)

plt.title("PCA Biplot of NBA Team Statistics")
plt.xlabel(f"PC1 ({pcaModel.explained_variance_ratio_[0] * 100:.1f}% variance)")
plt.ylabel(f"PC2 ({pcaModel.explained_variance_ratio_[1] * 100:.1f}% variance)")
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)
plt.tight_layout()
plt.show()

pc1Variance = pcaModel.explained_variance_ratio_[0] * 100
pc2Variance = pcaModel.explained_variance_ratio_[1] * 100

print(f"Analysis: PCA reduces the team stats into two main patterns, with PC1 explaining {pc1Variance:.1f}% of the variation and PC2 explaining {pc2Variance:.1f}%.")
```

### Principal Component Analysis (PCA)

PCA reduces multiple correlated team statistics into a smaller number of principal components, helping us identify the main patterns that distinguish successful from unsuccessful teams. By combining variables like offensive rating, defensive rating, pace, and roster age, PCA reveals which dimensions matter most for understanding team performance. PC1 and PC2 together explain over 77% of the variation in team statistics, suggesting that team success can be largely understood through two main patterns.

## Analysis

### Analysis Part 1: What Drives Team Success?

Our EDA establishes that team success in the NBA is dominated by a single underlying quality dimension, captured by PC1 (which combines SRS, offensive/defensive efficiency). This dimension explains 94% of variation in wins and 91% of variation in playoff seeding. 

This finding has critical implications for our research question: **Any association between strategic rest and team success may partly reflect that strong teams (high PC1) both rest more and win more due to superior talent.** Our control variable strategy using SRS, offensive rating, and defensive rating as controls is therefore essential. Without controlling for team strength, we cannot isolate whether rest independently predicts success or merely correlates with it.

Roster age and pace, while interesting, show weak direct relationships with wins. This validates our choice to include them as controls without expecting them to be primary drivers.

### Analysis Part 2: Pre/Post Policy Baseline Changes

Between the pre-policy era (2017-18 to 2022-23) and post-policy era (2023-24 to 2024-25), offensive rating increased significantly (p = 0.000) and defensive rating increased significantly (p = 0.000). However, pace showed no significant change (p = 0.248), and roster age showed no significant change (p = 0.492).

These findings suggest:
1. **The 2023 policy did not fundamentally alter how teams construct rosters or play at different speeds.** If the policy drove behavioral changes in rest patterns, those changes occurred without requiring teams to shift roster composition or game pace.
2. **Offensive and defensive rating increases reflect broader league evolution** (three-point emphasis, rule changes, evolved tactics) rather than policy-driven changes in load management strategy.
3. **Any detected policy effect on rest must come from changes in rest decisions themselves, not from secondary changes in team construction.** This is methodologically cleaner for Dataset 2 analysis.

### Analysis Part 3: Team Quality as the Baseline Model

PC1 serves as a nearly perfect predictor of team success (r = 0.94 with win %), leaving limited residual variance for other factors to explain. When we eventually introduce rest patterns from Dataset 2, the critical question becomes: **Do rest decisions add explanatory power beyond team strength?**

Put differently:
- **If rest adds no value:** Teams that rest more simply happen to be stronger teams (confounding).
- **If rest adds value:** Better teams achieve better results partly because they strategically rest while maintaining competitiveness.
- **If rest subtracts value:** Strong teams could achieve even better results by resting less (load management is counterproductive).

Our baseline model team strength predicts wins is extremely strong. Rest patterns must overcome this strong baseline to show meaningful independent effects.

### Narrow vs. Broad Rest Detection

The NBA's 2023 Player Participation Policy fines teams for resting star players in nationally televised games. This creates an incentive to reclassify strategic rest as minor injury. We construct two rest definitions to detect this:
- **Narrow**: explicit keywords — *rest, load management, not injury related, precautionary*
- **Broad**: narrow + vague single-session notes — *soreness, tightness, fatigue, maintenance*

A widening gap between the two post-2023 is consistent with teams substituting explicit rest labels for vague injury language.

```python
narrow_keywords = ['rest', 'load management', 'not injury related', 'precautionary']
broad_keywords = narrow_keywords + ['soreness', 'sore', 'tightness', 'tight', 'fatigue', 'maintenance', 'stiffness']

def flag_rest(notes, keywords):
    if pd.isna(notes):
        return False
    notes_lower = notes.lower()
    return any(kw in notes_lower for kw in keywords)

relinquished = relinquished.copy()
relinquished['narrow_rest'] = relinquished['notes'].apply(lambda x: flag_rest(x, narrow_keywords))
relinquished['broad_rest'] = relinquished['notes'].apply(lambda x: flag_rest(x, broad_keywords))

rest_by_season = relinquished.groupby('season').agg(
    total=('narrow_rest', 'count'),
    narrow=('narrow_rest', 'sum'),
    broad=('broad_rest', 'sum')
).reset_index()
rest_by_season['narrow_rate'] = rest_by_season['narrow'] / rest_by_season['total'] * 100
rest_by_season['broad_rate'] = rest_by_season['broad'] / rest_by_season['total'] * 100

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(rest_by_season['season'], rest_by_season['narrow_rate'], marker='o', label='Narrow (explicit rest)', color='#2c7bb6', linewidth=2)
ax.plot(rest_by_season['season'], rest_by_season['broad_rate'], marker='s', label='Broad (+ vague injury terms)', color='#d7191c', linewidth=2, linestyle='--')
ax.axvline(2024, linestyle=':', color='gray', alpha=0.8, label='2023 Policy effective')
ax.set_title('Rest Detection Rate by Season: Narrow vs. Broad Definition', fontsize=13)
ax.set_xlabel('Season (End Year)', fontsize=11)
ax.set_ylabel('% of Relinquish Events Flagged as Rest', fontsize=11)
ax.legend()
plt.tight_layout()
plt.show()

print(rest_by_season[['season','narrow_rate','broad_rate']].to_string(index=False))
```

The narrow rest rate is near zero across almost all seasons, indicating that teams almost never 
use explicit rest language in transaction notes — even before the 2023 policy. The only exception 
is 2017-18 (5.3%), which likely reflects an early period before teams became aware of the 
reputational risk of explicit rest labeling. The broad rate by contrast stays consistently 
elevated at 13–20%, reflecting the steady use of vague injury language like soreness and 
tightness throughout the window. This near-zero narrow rate is itself a key finding: it means 
our rest proxy must rely on the broad definition, and that the 2023 policy did not meaningfully 
change an already-established pattern of avoiding explicit rest labels.

### Absence Duration Distribution

The dataset records when players are placed on and activated from the inactive list, but does not directly state how many games were missed. We approximate absence duration by computing the gap in days between paired relinquish and acquire events for the same player and team within a season.

```python
relinquish_events = df2[df2['event_type'] == 'relinquished'][['date','team','player','season']].copy()
acquire_events = df2[df2['event_type'] == 'acquired'][['date','team','player','season']].copy()

relinquish_events = relinquish_events.sort_values('date')
acquire_events = acquire_events.sort_values('date')

paired = pd.merge_asof(
    relinquish_events,
    acquire_events.rename(columns={'date': 'return_date'}),
    left_on='date', right_on='return_date',
    by=['team', 'player', 'season'],
    direction='forward'
)

paired['gap_days'] = (paired['return_date'] - paired['date']).dt.days
paired_valid = paired.dropna(subset=['gap_days'])
paired_valid = paired_valid[paired_valid['gap_days'] > 0]

fig, ax = plt.subplots(figsize=(10, 5))
plot_data = paired_valid[paired_valid['gap_days'] <= 60]['gap_days']
ax.hist(plot_data, bins=30, color='#5b9bd5', edgecolor='white')
ax.set_title('Distribution of Absence Duration (Gap Days)\nTruncated at 60 Days for Readability', fontsize=13)
ax.set_xlabel('Days Between Relinquish and Acquire Events', fontsize=11)
ax.set_ylabel('Number of Events', fontsize=11)
ax.axvline(paired_valid['gap_days'].median(), color='#d9534f', linestyle='--', label=f"Median: {paired_valid['gap_days'].median():.0f} days")
ax.legend()
plt.tight_layout()
plt.show()

q1 = paired_valid['gap_days'].quantile(0.25)
q3 = paired_valid['gap_days'].quantile(0.75)
print(f"Median gap: {paired_valid['gap_days'].median():.0f} days")
print(f"IQR: {q1:.0f}–{q3:.0f} days")
print(f"Events > 30 days (likely true injuries): {(paired_valid['gap_days'] > 30).sum()}")
print(f"Events <= 3 days (likely rest): {(paired_valid['gap_days'] <= 3).sum()}")
```

The distribution is strongly right-skewed, so median and IQR are the appropriate summary statistics here. Short absences (1–3 days) form the dominant cluster and are the most likely candidates for strategic rest. The long right tail represents season-ending or multi-week injuries. Full games-missed calculation requires joining with the game schedule from Dataset #1, which will be completed at the analysis stage.

### Player Name Normalization: Fixing Accent Character Mismatches

Before rebuilding the analysis table, we address a data quality issue 
discovered during the matching process. Basketball Reference (Dataset #3) 
stores player names with their original accent characters — for example 
Bogdan Bogdanović, Dario Šarić, and Alperen Şengün. ProSportsTransactions 
(Dataset #2) strips those accents and writes Bogdan Bogdanovic, Dario Saric, 
and Alperen Sengun. These are the same players but Python string matching 
treats them as different names, causing real injury and rest events to be 
silently dropped during the join.

We fix this by applying Unicode NFD normalization to all player names in 
both datasets, which decomposes accented characters into their base letter 
plus a combining accent mark, then strips the accent marks. After 
normalization, both datasets use plain ASCII names and matching works 
correctly.

This recovered 13 additional players, increasing matched top-3 players from 
251 to 264. The 26 still-unmatched players have no accent characters — they 
are genuinely absent from the Dataset #2 injury log, either because they 
remained healthy throughout their top-3 seasons or their absences were not 
logged by ProSportsTransactions.

```python
import unicodedata

def normalize_name(name):
    if pd.isna(name): return name
    # Remove accents: converts ć → c, š → s, etc.
    return ''.join(
        c for c in unicodedata.normalize('NFD', str(name))
        if unicodedata.category(c) != 'Mn'
    )

# Normalize both sides
top3['Player_norm'] = top3['Player'].apply(normalize_name)
df2['player_norm']  = df2['player'].apply(normalize_name)

# Re-check match rate
top3_norm = set(top3['Player_norm'].unique())
df2_norm  = set(df2['player_norm'].dropna().unique())

matched_norm   = top3_norm & df2_norm
unmatched_norm = top3_norm - df2_norm

print(f"After accent normalization:")
print(f"Matched:   {len(matched_norm)}")
print(f"Unmatched: {len(unmatched_norm)}")
print(f"\nStill unmatched:")
print(sorted(unmatched_norm)[:20])
```

### Rebuilding the Analysis Table with Normalized Names

With normalization applied to both datasets, we rebuild the complete 
analysis table from scratch using the corrected player name matching. This 
process has three steps.

First, we filter Dataset #2's relinquish events to only those belonging to 
top-3 minutes leaders identified in Dataset #3, now using normalized names 
as the join key. This increases matched events from 1,365 to 1,451 and 
broad rest events from 291 to 314.

Second, we aggregate those events to the team-season level to compute 
rest_rate — the proportion of a team's top-3 player absences that were 
classified as broad rest.

Third, we merge the aggregated rest rates with Dataset #1's team-season 
outcomes using the shared (team_abbr, season) key, producing the final 
240-row analysis table. Teams with no matched top-3 absences receive a 
rest_rate of 0. The corrected table has 138 team-seasons with rest_rate > 0, 
up from 132 before normalization.

```python
# Rebuild top3_set using normalized names
top3_set_norm = set(zip(top3['Player_norm'], top3['Team'], top3['season']))

relinquish = df2[df2['event_type'] == 'relinquished'].copy()
relinquish['broad_rest'] = relinquish['notes'].apply(is_rest)
relinquish['is_top3_player'] = relinquish.apply(
    lambda r: (r['player_norm'], r['team_abbr'], r['season']) in top3_set_norm, axis=1
)

top3_events = relinquish[relinquish['is_top3_player']].copy()
print(f"Events matching top-3 players (before): 1365")
print(f"Events matching top-3 players (after):  {len(top3_events)}")
print(f"Broad rest events: {top3_events['broad_rest'].sum()}")

# Rebuild rest_agg and analysis table
rest_agg = (
    top3_events
    .groupby(['team_abbr', 'season'])
    .agg(
        total_top3_absences=('broad_rest', 'count'),
        rest_absences=('broad_rest', 'sum')
    )
    .reset_index()
)
rest_agg['rest_rate'] = rest_agg['rest_absences'] / rest_agg['total_top3_absences']

analysis = df1[['team', 'team_abbr', 'season', 'win_pct', 'srs', 'roster_age', 'playoff_seed']].merge(
    rest_agg,
    on=['team_abbr', 'season'],
    how='left'
)
analysis['rest_rate']           = analysis['rest_rate'].fillna(0)
analysis['total_top3_absences'] = analysis['total_top3_absences'].fillna(0)
analysis['rest_absences']       = analysis['rest_absences'].fillna(0)

print(f"\nUpdated analysis table: {analysis.shape}")
print(f"Teams with rest_rate > 0: {(analysis['rest_rate'] > 0).sum()}")

# Save
analysis.to_csv('data/02-processed/nba_analysis_final.csv', index=False)
print("Saved.")
```

## Experiment 1: Raw Correlation Between Rest Rate and Team Outcomes

We begin with the simplest possible question: before adding any controls, 
is there any raw linear relationship between how much a team rests its 
top-3 minutes leaders and how well that team performs? We compute Pearson 
correlation coefficients between rest_rate and two outcome variables, win 
percentage and playoff seed, and visualize the relationship using scatter 
plots with regression lines.

This step is intentionally done without control variables. We want to see 
the raw picture first, because understanding whether and how it changes 
when controls are added is itself an important part of the analysis.

However, before interpreting the full-window results, we must account for 
a data quality issue identified in the EDA. The Dataset #2 notes field 
had a high rate of non-specific entries in 2018-2019, with 44% and 28% of 
absences respectively providing no classifiable reason. Since our rest 
classifier relies on keyword matching in the notes field, these entries 
are invisible to it and get counted as zero rest regardless of what 
actually happened. This creates artificially suppressed rest rates in the 
early seasons specifically, which can produce spurious correlations.

We therefore run the analysis on two windows. The full window uses all 240 
team-seasons from 2018-2025. The restricted window uses only the 150 
team-seasons from 2021-2025, where non-specific note rates dropped below 
10% and rest classification is reliable. This sensitivity check was 
pre-committed in the EDA checkpoint. If the full-window correlation holds 
in the restricted window, the finding is robust. If it disappears, the 
full-window result was a reporting artifact.

Note that the vertical cluster of points at rest_rate = 0 in both scatter 
plots reflects two overlapping groups: teams whose top-3 players genuinely 
had no logged absences in the injury data, and teams where absences existed 
but could not be classified due to non-specific notes. This cluster 
illustrates the data quality limitation directly.

```python
# ================================================================
# EXPERIMENT 1: Raw Correlation — Rest Rate vs Team Outcomes
# ================================================================
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

analysis = pd.read_csv('data/02-processed/nba_analysis_final.csv')
print(f"Analysis table: {analysis.shape}")
print(f"Teams with rest_rate > 0: {(analysis['rest_rate'] > 0).sum()}")

# ── Full window plots ─────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.regplot(data=analysis, x='rest_rate', y='win_pct',
            ax=axes[0], scatter_kws={'alpha': 0.5, 'color': 'steelblue'})
r1, p1 = stats.pearsonr(analysis['rest_rate'], analysis['win_pct'])
axes[0].set_title(f'Rest Rate vs Win Percentage (Full Window)\nr = {r1:.3f}, p = {p1:.3f}')
axes[0].set_xlabel('Broad Rest Rate (Top-3 Players)')
axes[0].set_ylabel('Win Percentage')

sns.regplot(data=analysis, x='rest_rate', y='playoff_seed',
            ax=axes[1], scatter_kws={'alpha': 0.5, 'color': 'coral'})
r2, p2 = stats.pearsonr(analysis['rest_rate'], analysis['playoff_seed'])
axes[1].set_title(f'Rest Rate vs Playoff Seed (Full Window)\nr = {r2:.3f}, p = {p2:.3f}')
axes[1].set_xlabel('Broad Rest Rate (Top-3 Players)')
axes[1].set_ylabel('Playoff Seed (lower = better)')
axes[1].invert_yaxis()

plt.suptitle('Raw Relationship: Rest Rate and Team Success (2017-18 to 2024-25)',
             fontsize=13, y=1.02)
plt.tight_layout()
plt.show()

# ── Sensitivity check: restricted to 2021-2025 ───────────────────
analysis_restricted = analysis[analysis['season'] >= 2021].copy()

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.regplot(data=analysis_restricted, x='rest_rate', y='win_pct',
            ax=axes[0], scatter_kws={'alpha': 0.5, 'color': 'steelblue'})
r1r, p1r = stats.pearsonr(analysis_restricted['rest_rate'],
                           analysis_restricted['win_pct'])
axes[0].set_title(f'Rest Rate vs Win Pct (2021-2025 only)\nr = {r1r:.3f}, p = {p1r:.3f}')
axes[0].set_xlabel('Broad Rest Rate (Top-3 Players)')
axes[0].set_ylabel('Win Percentage')

sns.regplot(data=analysis_restricted, x='rest_rate', y='playoff_seed',
            ax=axes[1], scatter_kws={'alpha': 0.5, 'color': 'coral'})
r2r, p2r = stats.pearsonr(analysis_restricted['rest_rate'],
                           analysis_restricted['playoff_seed'])
axes[1].set_title(f'Rest Rate vs Playoff Seed (2021-2025 only)\nr = {r2r:.3f}, p = {p2r:.3f}')
axes[1].set_xlabel('Broad Rest Rate (Top-3 Players)')
axes[1].set_ylabel('Playoff Seed (lower = better)')
axes[1].invert_yaxis()

plt.suptitle('Sensitivity Check: Restricted to Seasons with Reliable Note Quality',
             fontsize=13, y=1.02)
plt.tight_layout()
plt.show()

# ── Summary ───────────────────────────────────────────────────────
print("\n" + "="*55)
print("EXPERIMENT 1 SUMMARY")
print("="*55)
print(f"\nFull window (2018-2025, n=240):")
print(f"  rest_rate vs win_pct:      r={r1:.3f}, p={p1:.3f}")
print(f"  rest_rate vs playoff_seed: r={r2:.3f}, p={p2:.3f}")
print(f"\nRestricted window (2021-2025, n={len(analysis_restricted)}):")
print(f"  rest_rate vs win_pct:      r={r1r:.3f}, p={p1r:.3f}")
print(f"  rest_rate vs playoff_seed: r={r2r:.3f}, p={p2r:.3f}")
```

**Full window (2018-2025, n=240):**

- rest_rate vs playoff_seed: r=-0.171, p=0.008

**Restricted window (2021-2025, n=150):**

The full window produces statistically significant correlations on both

However, both correlations collapse entirely in the restricted window. The

We therefore treat the restricted window as our primary sample and 
conclude that there is no meaningful raw correlation between a team's 
broad rest rate among top-3 minute leaders and either win percentage or

null result holds after formally controlling for team quality.

## Experiment 2: OLS Regression — Primary Research Question

Experiment 1 showed that the raw correlation between rest rate and team 
outcomes disappears in the reliable data window. Experiment 2 tests whether 
this holds formally using OLS regression, and adds a critical methodological 
step: controlling for team quality.

The central confound in any rest-performance analysis is that good teams have 
star players worth resting, and good teams also win more. This means rest rate 
and win percentage could both be caused by a third variable — team quality — 
rather than rest causing wins. To isolate any independent effect of rest, we 
must control for team strength before drawing any conclusions.

We run three models on both the full window (2018-2025, n=240) and the 
restricted reliable window (2021-2025, n=150):

- **Model A** — win_pct ~ rest_rate only. No controls. This gives the raw 
  regression equivalent of Experiment 1's correlation.
- **Model B** — win_pct ~ rest_rate + srs + roster_age. SRS (Simple Rating 
  System) captures overall team quality on a points-per-game scale. Roster 
  age captures experience effects. This is the primary model for answering 
  the research question.
- **Model C** — playoff_seed ~ rest_rate + srs + roster_age. Same controls 
  applied to the seeding outcome variable.

The bar chart compares the rest_rate coefficient across all four 
specifications — full vs restricted window, with and without controls — to 
visualize how the effect changes as we add rigor. Red bars indicate 
statistically significant coefficients (p < 0.05), grey bars are not 
significant.

```python
# ================================================================
# EXPERIMENT 2: OLS Regression — Primary Research Question
# ================================================================
from statsmodels.formula.api import ols

# Run on both windows for consistency
for label, data in [('Full window (2018-2025)', analysis),
                    ('Restricted window (2021-2025)', analysis_restricted)]:

    print(f"\n{'='*60}")
    print(f"{label}  n={len(data)}")
    print('='*60)

    # Model A: no controls
    A = ols('win_pct ~ rest_rate', data=data).fit()

    # Model B: with controls
    B = ols('win_pct ~ rest_rate + srs + roster_age', data=data).fit()

    # Model C: playoff seed with controls
    C = ols('playoff_seed ~ rest_rate + srs + roster_age', data=data).fit()

    print(f"\nModel A — win_pct ~ rest_rate only:")
    print(f"  R²={A.rsquared:.4f}  |  rest_rate coef={A.params['rest_rate']:.4f}  |  p={A.pvalues['rest_rate']:.4f}")

    print(f"\nModel B — win_pct ~ rest_rate + srs + roster_age:")
    print(f"  R²={B.rsquared:.4f}  |  rest_rate coef={B.params['rest_rate']:.4f}  |  p={B.pvalues['rest_rate']:.4f}")
    print(f"  srs coef={B.params['srs']:.4f}  p={B.pvalues['srs']:.4f}")
    print(f"  roster_age coef={B.params['roster_age']:.4f}  p={B.pvalues['roster_age']:.4f}")

    print(f"\nModel C — playoff_seed ~ rest_rate + srs + roster_age:")
    print(f"  R²={C.rsquared:.4f}  |  rest_rate coef={C.params['rest_rate']:.4f}  |  p={C.pvalues['rest_rate']:.4f}")
    print(f"  srs coef={C.params['srs']:.4f}  p={C.pvalues['srs']:.4f}")

# Visual: compare rest_rate coefficient before and after controls (restricted window)
B_full = ols('win_pct ~ rest_rate + srs + roster_age', data=analysis).fit()
B_rest = ols('win_pct ~ rest_rate + srs + roster_age', data=analysis_restricted).fit()
A_full = ols('win_pct ~ rest_rate', data=analysis).fit()
A_rest = ols('win_pct ~ rest_rate', data=analysis_restricted).fit()

fig, ax = plt.subplots(figsize=(10, 5))
labels  = ['Full\n(no controls)', 'Full\n(with controls)',
           'Restricted\n(no controls)', 'Restricted\n(with controls)']
coefs   = [A_full.params['rest_rate'], B_full.params['rest_rate'],
           A_rest.params['rest_rate'],  B_rest.params['rest_rate']]
pvals   = [A_full.pvalues['rest_rate'], B_full.pvalues['rest_rate'],
           A_rest.pvalues['rest_rate'],  B_rest.pvalues['rest_rate']]
colors  = ['#d9534f' if p < 0.05 else '#aaaaaa' for p in pvals]

bars = ax.bar(labels, coefs, color=colors, edgecolor='black', width=0.5)
ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
ax.set_ylabel('rest_rate Coefficient')
ax.set_title('Rest Rate Effect on Win Percentage\nAcross Windows and Model Specifications\n(Red = p<0.05, Grey = not significant)')

for bar, coef, p in zip(bars, coefs, pvals):
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height() + 0.001,
            f'coef={coef:.3f}\np={p:.3f}',
            ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

**Full window (2018-2025, n=240):**

Model A shows rest_rate is significant without controls (coef=0.090, p=0.039, 
R²=0.018), meaning rest rate alone appears to predict win percentage. However 
R²=0.018 means it explains less than 2% of the variation in wins — already a 
sign the effect is weak.

The moment controls are added in Model B, the rest_rate coefficient shrinks 
from 0.090 to 0.018 and becomes non-significant (p=0.136), while R² jumps 
from 0.018 to 0.929. SRS alone explains nearly all the variation in win 
percentage (coef=0.0288, p<0.001). This is a textbook confounding pattern: 
the apparent relationship between rest and winning was driven entirely by the 
fact that strong teams (high SRS) both win more and rest more. Once team 
quality is held constant, rest has no independent predictive power.

Model C shows a significant association between rest rate and playoff seeding 
in the full window (coef=-2.756, p=0.004), meaning higher rest rates are 
associated with better seeds. However this full-window result is subject to 
the same data quality limitations identified in Experiment 1.

**Restricted window (2021-2025, n=150):**

Model A confirms the Experiment 1 finding — no raw relationship (coef=0.020, 
p=0.763, R²=0.0006). Rest rate alone explains essentially 0% of variation 
in win percentage in reliable data.

Model B shows the rest_rate coefficient is effectively zero after controls 
(coef=0.001, p=0.966), while R²=0.932. SRS again dominates (p<0.001) with 
roster age showing a small but significant positive effect (coef=0.0047, 
p=0.021), suggesting experienced rosters win slightly more after controlling 
for overall team quality.

Model C in the restricted window shows the playoff seed association is now 
borderline non-significant (coef=-2.905, p=0.053), falling just above the 
0.05 threshold. Three limitations from the EDA are directly relevant here: 
post-2024 mislabeling of rest as injury likely suppresses rest_rate for the 
best-seeded teams in the post-policy seasons, the 2025 data is incomplete, 
and 26 top-3 players were unmatched in the injury log creating genuine zeros 
that may not reflect actual rest behavior.

**The bar chart summarizes the complete picture.** Only one bar is red — the 
full window without controls. Every other specification is grey. This 
visually confirms that any apparent effect of rest on winning is eliminated 
the moment either reliable data or appropriate controls are applied.

**Answer to the primary research question:** After controlling for team 
strength (SRS) and roster age, there is no statistically significant 
association between the broad rest rate among a team's top-3 minutes leaders 
and regular season win percentage. The R²=0.932 of the restricted Model B 
confirms that team quality as measured by SRS explains virtually all 
variation in wins, leaving no meaningful independent role for rest behavior. 
This finding is consistent with Belk et al. (2017) and Pradhan and Miller 
(2022), both of whom found no significant performance benefit from strategic 
rest in prior NBA seasons.

## Experiment 3: Pre/Post Policy Analysis — Secondary Research Question

The 2023 NBA Player Participation Policy, which took effect at the start 
of the 2023-24 season, introduced financial penalties for teams that rest 
healthy star players in high-profile games. The policy specifically targets 
players defined as stars — those who made an All-Star or All-NBA team in 
any of the three prior seasons. Our secondary research question asks whether 
this policy actually changed rest behavior, and specifically whether it 
changed rest behavior differently for stars versus non-star top-3 players.

The design of this experiment follows a difference-in-differences logic. 
If the policy worked, we would expect star players (who face the new 
penalties) to show a decline in classified rest events after 2023, while 
non-star top-3 players (who are not subject to the policy's fines) should 
show no meaningful change. The contrast between these two groups is the 
test.

We restrict to the reliable window (2021-2025) throughout this experiment 
to avoid the data quality issues from 2018-2019. This gives us three 
pre-policy seasons (2021, 2022, 2023) and two post-policy seasons (2024, 
2025), with 2025 being incomplete due to the Dataset #2 snapshot ending 
January 12, 2025.

There are two critical limitations that apply to the entire experiment and 
must be kept in mind when interpreting every result.

First, the post-2023 mislabeling problem. Teams now have a financial 
incentive to label strategic rest as minor injury rather than rest, to 
avoid fines. Our broad keyword classifier catches terms like soreness and 
tightness, but teams using specific injury language to disguise rest would 
evade detection. This means post-2024 rest rates are likely systematically 
undercounted for stars specifically, since they are the most valuable 
players and the ones most worth protecting through relabeling. Any decline 
in star rest rates after the policy could therefore reflect either genuine 
behavioral change or improved disguise, and the data cannot distinguish 
between these two explanations.

Second, the small post-policy sample. With only 1.5 effective post-policy 
seasons, statistical power is limited and results should be treated as 
preliminary.

**Step 1** builds the star eligibility flag using the exact definition from 
the policy: a player is policy-eligible in season S if they received an 
All-Star selection or an All-NBA First, Second, or Third Team selection in 
any of the three prior seasons (S-1, S-2, or S-3). We parse the Awards 
column from Dataset #3, which stores award codes as concatenated strings 
such as ASNBA1 or MVP-3ASNBA2.

```python
# ================================================================
# EXPERIMENT 3: Pre/Post Policy — Secondary Research Question
# ================================================================
# Did the 2023 Player Participation Policy change rest patterns
# for policy-eligible stars vs non-star top-3 players?
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# ── Load everything fresh with normalized names ───────────────────
df2   = pd.read_csv('data/02-processed/nba_injury_events_clean.csv')
top3  = pd.read_csv('data/02-processed/nba_top3_leaders_by_team_season.csv')
df_full = pd.read_csv('data/02-processed/nba_player_stats_clean.csv')

import unicodedata
def normalize_name(name):
    if pd.isna(name): return name
    return ''.join(
        c for c in unicodedata.normalize('NFD', str(name))
        if unicodedata.category(c) != 'Mn'
    )

df2['player_norm']   = df2['player'].apply(normalize_name)
top3['Player_norm']  = top3['Player'].apply(normalize_name)
df_full['Player_norm'] = df_full['Player'].apply(normalize_name)

# ── Step 1: Build star eligibility (rolling 3-season window) ─────
# A player is policy-eligible in season S if they received
# All-Star (AS) or All-NBA (NBA1/NBA2/NBA3) in any of S-1, S-2, S-3
# We parse the Awards column carefully using substrings

def has_star_award(awards_str):
    if pd.isna(awards_str): return False
    a = str(awards_str)
    # Check for AS (All-Star) - careful not to match 'SAS' or other substrings
    # NBA1, NBA2, NBA3 are unambiguous
    has_as   = 'AS' in a and not a.replace('AS','').startswith('S')
    has_nba  = any(x in a for x in ['NBA1','NBA2','NBA3'])
    return has_as or has_nba

# Build season-level award lookup
award_lookup = (
    df_full
    .groupby(['Player_norm','season'])['Awards']
    .first()
    .reset_index()
)
award_lookup['had_award'] = award_lookup['Awards'].apply(has_star_award)

def is_star_eligible(player_norm, season, lookup):
    prior = lookup[
        (lookup['Player_norm'] == player_norm) &
        (lookup['season'].isin([season-1, season-2, season-3]))
    ]
    return bool(prior['had_award'].any())

# Apply to top-3 table
print("Building star eligibility (this may take a moment)...")
top3['star_eligible'] = top3.apply(
    lambda r: is_star_eligible(r['Player_norm'], r['season'], award_lookup),
    axis=1
)

print("\nStar-eligible top-3 players per season:")
print(top3.groupby('season')['star_eligible'].sum().to_string())
print(f"\nTotal star-eligible player-team-seasons: {top3['star_eligible'].sum()}")
print(f"Total non-star top-3 player-team-seasons: {(~top3['star_eligible']).sum()}")

# Sanity check — these names should appear as stars
sample_stars = top3[top3['star_eligible']][['Player','Team','season']].head(10)
print("\nSample star-eligible players:")
print(sample_stars.to_string(index=False))
```

### Star Eligibility Results

The star eligibility counts look correct and pass several sanity checks.

Season 2018 shows zero eligible stars, which is expected and correct. The 
rolling window requires award data from seasons 2015, 2016, and 2017, which 
fall outside our dataset. This is a structural limitation of the first 
season rather than a coding error, and it is one reason we use 2021 as our 
pre-policy start rather than 2018.

From 2019 onward the counts stabilize at roughly 18-29 star-eligible 
players per season, representing about 20-30% of all 90 top-3 player 
slots available each season (30 teams x 3 players). This proportion is 
plausible — not every team has a player with recent All-Star or All-NBA 
recognition as one of their top-3 minutes leaders.

The sample star-eligible players for 2019 validate the logic correctly. 
Kyrie Irving, Kevin Durant, Stephen Curry, LeBron James, Giannis 
Antetokounmpo, and James Harden are all legitimate superstars who held 
prior All-Star or All-NBA selections. Kemba Walker made the All-Star team 
in 2019 itself but would not qualify until 2020 under our rolling window. 
His appearance here means he had a qualifying selection in 2016, 2017, or 
2018, which is also accurate — he was an All-Star in 2017 and 2018.

One note on Andre Drummond appearing as a star in 2019. He made the 
All-NBA Third Team in 2016, which falls within the three-season lookback 
window. Whether a player of his caliber constitutes a "star" in the 
colloquial sense is debatable, but under the strict policy definition he

Across the full dataset, 166 player-team-seasons are star-eligible and 554 
are not, giving us a reasonable sample size in both groups for the 
pre/post comparison.

### Flagging Rest Events for Stars vs Non-Stars

With star eligibility established, we now filter Dataset #2's injury events 
to identify rest events specifically for top-3 players, and label each event 
as belonging to a star or non-star player.

We use the same broad keyword classifier applied throughout the analysis, 
matching on terms including rest, load management, soreness, sore, tightness, 
tight, fatigue, maintenance, stiffness, precautionary, and not injury related. 
This is the broad definition used in all prior experiments rather than the 
narrow explicit-rest definition, because the narrow rate has been near zero 
across all seasons — teams rarely use the word "rest" directly even before 
the policy existed.

We then restrict to the reliable window of 2021-2025, split at the 2024 
season boundary where the policy took effect. This gives us two groups 
(stars and non-stars) across two periods (pre-policy 2021-2023 and 
post-policy 2024-2025), producing the four cells of the comparison.

One structural asymmetry in the counts is expected and worth flagging 
upfront. The pre-policy window spans three seasons and the post-policy 
window spans roughly 1.5 seasons (2024 is complete, 2025 ends January 12). 
This means the post-policy counts will naturally be smaller than the 
pre-policy counts for both groups, not because fewer rest events occurred 
per season but simply because fewer games were played in the post-policy 
window. We account for this by comparing rest rates (proportions) rather 
than raw event counts.

```python
# ── Step 2: Flag rest events for stars vs non-stars ───────────────

broad_keywords = [
    'rest', 'load management', 'not injury related', 'precautionary',
    'soreness', 'sore', 'tightness', 'tight', 'fatigue',
    'maintenance', 'stiffness'
]

def is_rest(note):
    if pd.isna(note): return False
    return any(kw in note.lower() for kw in broad_keywords)

relinquish = df2[df2['event_type'] == 'relinquished'].copy()
relinquish['broad_rest'] = relinquish['notes'].apply(is_rest)

# Build lookup sets using normalized names
top3_set = set(zip(top3['Player_norm'], top3['Team'], top3['season']))
star_set  = set(zip(
    top3[top3['star_eligible']]['Player_norm'],
    top3[top3['star_eligible']]['Team'],
    top3[top3['star_eligible']]['season']
))

relinquish['is_top3'] = relinquish.apply(
    lambda r: (r['player_norm'], r['team_abbr'], r['season']) in top3_set, axis=1
)
relinquish['is_star'] = relinquish.apply(
    lambda r: (r['player_norm'], r['team_abbr'], r['season']) in star_set, axis=1
)

top3_events = relinquish[relinquish['is_top3']].copy()

# Policy boundary: 2024+ is post-policy
top3_events['post_policy'] = top3_events['season'] >= 2024

# IMPORTANT LIMITATION FLAG:
# We use 2021-2025 as our window to exclude the
# unreliable 2018-2019 early seasons (44%/28% non-specific notes)
# Pre-policy: 2021-2023 (reliable, pre-policy)
# Post-policy: 2024-2025 (policy in effect, but rest may be mislabeled)

top3_reliable = top3_events[top3_events['season'] >= 2021].copy()

print(f"Top-3 events in reliable window (2021-2025): {len(top3_reliable)}")
print(f"\nBreakdown:")
print(f"  Stars pre-policy  (2021-2023): {len(top3_reliable[(top3_reliable['is_star']) & (~top3_reliable['post_policy'])])}")
print(f"  Stars post-policy (2024-2025): {len(top3_reliable[(top3_reliable['is_star']) & (top3_reliable['post_policy'])])}")
print(f"  Non-stars pre-policy:          {len(top3_reliable[(~top3_reliable['is_star']) & (~top3_reliable['post_policy'])])}")
print(f"  Non-stars post-policy:         {len(top3_reliable[(~top3_reliable['is_star']) & (top3_reliable['post_policy'])])}")
```

### Event Breakdown Results

The 1,039 total events in the reliable window break down as expected across 
the four cells.

Stars have 278 pre-policy events and 141 post-policy events. The ratio of 
roughly 2:1 is consistent with having three pre-policy seasons versus 
approximately 1.5 post-policy seasons. Non-stars show the same pattern with 
455 pre-policy and 165 post-policy events, again roughly a 2.75:1 ratio 
reflecting the same seasonal imbalance.

Importantly, non-stars have substantially more total events than stars 
across both periods. This makes sense — there are roughly 3.3 times as many 
non-star top-3 player-team-seasons (554) as star ones (166), so we would 
expect non-stars to generate more total absence events.

These sample sizes are sufficient for t-tests and the difference-in-differences 
model in Steps 3 and 4, though the post-policy samples (141 stars, 165 
non-stars) are on the smaller side and statistical power will be limited. 
This is a structural constraint of the data — the policy has only been in 
effect for 1.5 seasons at the time of our snapshot — and is one reason we 
frame the secondary analysis as exploratory rather than confirmatory.

### Computing Rest Rates and Testing for Policy Effects

With events flagged by group and period, we now compute rest rates for each 
of the four cells and test whether the changes across periods are 
statistically significant.

Rest rate here is the proportion of absences classified as broad rest for 
each group in each period. We use individual t-tests to ask two separate 
questions: did star rest rates change significantly between pre and post 
policy, and did non-star rest rates change significantly? 

The non-star group serves as a natural control. Since the policy 
specifically targets star players, non-stars should not be affected by it. 
If non-stars show no meaningful change while stars show a decline, that 
pattern is consistent with a policy-driven effect rather than some 
league-wide shift that would have happened regardless.

We also report the raw direction and magnitude of change for each group, 
since effect size is as important as statistical significance when sample 
sizes are small.

```python
# ── Step 3: Compute rest rates and run t-tests ────────────────────

stars_pre    = top3_reliable[(top3_reliable['is_star'])  & (~top3_reliable['post_policy'])]['broad_rest'].astype(float)
stars_post   = top3_reliable[(top3_reliable['is_star'])  & (top3_reliable['post_policy'])]['broad_rest'].astype(float)
nonstars_pre = top3_reliable[(~top3_reliable['is_star']) & (~top3_reliable['post_policy'])]['broad_rest'].astype(float)
nonstars_post= top3_reliable[(~top3_reliable['is_star']) & (top3_reliable['post_policy'])]['broad_rest'].astype(float)

print("="*60)
print("REST RATES BY GROUP AND PERIOD (2021-2025 reliable window)")
print("="*60)
print(f"\nStars — Pre-policy  (2021-2023): {stars_pre.mean():.3f}  (n={len(stars_pre)})")
print(f"Stars — Post-policy (2024-2025): {stars_post.mean():.3f}  (n={len(stars_post)})")
print(f"\nNon-Stars — Pre-policy  (2021-2023): {nonstars_pre.mean():.3f}  (n={len(nonstars_pre)})")
print(f"Non-Stars — Post-policy (2024-2025): {nonstars_post.mean():.3f}  (n={len(nonstars_post)})")

t_star, p_star   = stats.ttest_ind(stars_pre, stars_post)
t_ns,   p_ns     = stats.ttest_ind(nonstars_pre, nonstars_post)

print(f"\nT-test Stars pre vs post:     t={t_star:.3f}, p={p_star:.3f}")
print(f"T-test Non-Stars pre vs post: t={t_ns:.3f},  p={p_ns:.3f}")

# Direction of change
star_change    = stars_post.mean()    - stars_pre.mean()
nonstar_change = nonstars_post.mean() - nonstars_pre.mean()
print(f"\nChange in rest rate after policy:")
print(f"  Stars:     {star_change:+.3f}")
print(f"  Non-Stars: {nonstar_change:+.3f}")
```

### Results and Interpretation

The results show a clear directional contrast between the two groups.

Stars declined from a pre-policy rest rate of 0.241 to 0.163 after the 
policy, a drop of 7.8 percentage points. The t-test for this change 
produces t=1.837, p=0.067 — directionally significant but falling just 
above the conventional 0.05 threshold.

Non-stars show essentially no change, moving from 0.204 to 0.212, an 
increase of just 0.8 percentage points in the opposite direction. The 
t-test for non-stars produces p=0.834, confirming that nothing meaningful 
happened to this group across the policy boundary.

The contrast between these two groups is the key pattern. Stars went down 
by 7.8 points while non-stars moved by less than 1 point in the other 
direction. This is exactly the differential effect you would expect if the 
policy influenced star rest behavior specifically, while leaving non-star 
behavior unchanged.

However, two important caveats apply before concluding the policy worked.

First, the p=0.067 for stars does not cross the 0.05 significance 
threshold. With only 141 post-policy star events representing roughly 1.5 
seasons of data, the test is underpowered. The borderline p-value is more 
likely a reflection of small sample size than of no real effect.

Second, and more fundamentally, the decline in star rest rate has two 
competing explanations that cannot be separated with this data. Teams may 
genuinely be resting stars less to avoid fines. Alternatively, teams may 
be resting stars at similar or higher rates but now calling those absences 
"left knee soreness" or "right hamstring tightness" rather than "rest", 
specifically to avoid the policy penalties. Our broad classifier catches 
soreness and tightness, but teams that adopt more specific injury language 
to disguise rest would evade it entirely. The sharp drop in star rest 
rates after 2023 is fully consistent with both interpretations, and the 
current data cannot determine which is driving the result.

We proceed to Step 4 to visualize these patterns and run the 
difference-in-differences model, which tests the contrast between groups 
more directly in a single statistical test.

### Difference-in-Differences Model

The t-tests in Step 3 tested each group separately, asking "did stars 
change?" and "did non-stars change?" as two independent questions. But the 
actual research question is a single question: did the policy affect stars 
differently than non-stars? These are not the same thing.

The difference-in-differences model answers the single correct question 
in one test. It estimates the interaction between group (star vs non-star) 
and period (pre vs post-policy), asking whether the change in stars was 
statistically distinguishable from the change in non-stars. This 
interaction term, labeled post:group in the output, is the DiD estimator 
and is the most rigorous single number we can produce for the secondary 
research question.

The model also provides an additional benefit. If some league-wide change 
happened after 2023 that affected all players regardless of star status — 
for example a shorter 2024-25 schedule or general changes in how teams 
manage fatigue — the non-star group absorbs that common trend. The DiD 
estimator then isolates only the differential effect on stars above and 
beyond whatever happened to non-stars, which is precisely the policy 
effect we are trying to measure.

```python
# Fix: convert broad_rest to integer before the model
top3_reliable['broad_rest_int'] = top3_reliable['broad_rest'].astype(int)
top3_reliable['group'] = top3_reliable['is_star'].astype(int)
top3_reliable['post']  = top3_reliable['post_policy'].astype(int)

did_model = smf.ols('broad_rest_int ~ post + group + post:group',
                    data=top3_reliable).fit()

print("="*60)
print("DIFFERENCE-IN-DIFFERENCES MODEL")
print("Outcome: broad_rest (1=rest event, 0=non-rest event)")
print("="*60)
print(f"\nBaseline (non-star, pre-policy):    {did_model.params['Intercept']:.3f}")
print(f"Post-policy effect (non-stars):     {did_model.params['post']:.3f}  p={did_model.pvalues['post']:.3f}")
print(f"Star effect (pre-policy):           {did_model.params['group']:.3f}  p={did_model.pvalues['group']:.3f}")
print(f"\nDiD estimator (post:group):         {did_model.params['post:group']:.3f}  p={did_model.pvalues['post:group']:.3f}")
print(f"\nInterpretation: Stars changed by {did_model.params['post:group']:.3f} more than non-stars after the policy.")
print(f"R² = {did_model.rsquared:.4f}")
```

### Difference-in-Differences Results and Final Data Interpretation

**Reading the output line by line:**

The baseline of 0.204 is simply the non-star pre-policy rest rate, 
confirming the Step 3 numbers are consistent.

The post-policy effect for non-stars is 0.008 with p=0.835, confirming 
that the control group showed essentially no change after the policy. This 
is exactly what a valid DiD design requires — if non-stars had shifted 
dramatically, it would suggest some confounding factor unrelated to the 
policy was driving the results.

The star effect pre-policy is 0.037 with p=0.238, meaning stars rested 
about 3.7 percentage points more than non-stars before the policy. This 
is not statistically significant but is directionally expected — teams 
rest their most valuable players more.

**The key number is the DiD estimator: -0.086, p=0.127.**

Stars declined by 8.6 percentage points more than non-stars after the 
policy. The direction is exactly what a policy effect would predict. The 
magnitude is practically meaningful — an 8.6 percentage point differential 
decline in rest classification represents a real shift in how star 
absences are recorded. However p=0.127 does not cross the conventional 
0.05 significance threshold, meaning we cannot formally reject the null 
hypothesis of no differential effect.

The R²=0.0034 confirms that group and policy period together explain very 
little of the raw variance in individual absence classification. This is 
expected — whether any individual absence is classified as rest depends 
heavily on the specific wording of the notes field, which introduces 
substantial noise at the event level.

**Final answer to the secondary research question:**

The 2023 Player Participation Policy was associated with a directional but 
statistically inconclusive differential reduction in classified rest events 
for policy-eligible star players relative to non-star top-3 players. The 
DiD estimate of -0.086 (p=0.127) indicates stars declined roughly 8.6 
percentage points more than non-stars after the policy, a pattern 
consistent with a behavioral response to the new fines. The time series 
shows a sharp inflection at the exact policy boundary for stars while 
non-stars remain flat throughout, which visually reinforces the 
directional story.

However, three factors prevent a definitive conclusion. First, the 
post-policy window covers only 1.5 effective seasons, limiting statistical 
power. A longer post-policy sample would likely resolve the borderline 
significance. Second, even if rest behavior genuinely changed, the

# Discussion and Conclusion

## What we set out to do

We started this project with a question that comes up every NBA season: does it actually help a team to rest its stars during the regular season? And underneath that, a more recent question - did the NBA's 2023 Player Participation Policy, which fines teams for sitting healthy stars in nationally televised games, actually change how teams behave? We combined three datasets (team stats from Basketball Reference, the ProSportsTransactions injury and rest event log, and per-game player stats also from Basketball Reference) to build a single team-season view of rest behavior alongside team outcomes from 2017–18 through 2024–25.

## What we found about the primary question
The cleanest answer in the project is also the most boring one: **once you control for team strength, strategic rest does not predict winning**. The first hint of this came from the EDA itself. The “Correlation Heatmap of NBA Team Statistics” and the “PCA Biplot of NBA Team Statistics” both show that team quality, captured by SRS, runs the show. SRS alone correlates with win percentage at roughly r = 0.96, and the first principal component (which is essentially "team quality") explains 94% of the variation in wins. This means that there is very little room left for anything else.

Experiment 1 then tested the rest-versus-winning relationship directly. There is a small positive correlation between broad rest rate and win percentage in the full window, but the moment we either restrict to the reliable 2021–2025 window or add SRS as a control, that correlation evaporates. 

Experiment 2 made this formal: in the Bar Chart of Rest Rate Coefficients With and Without Controls, the only bar that stays red (significant) is the one without controls, and even that one collapses in the restricted window. The R² jumping from about 2% (rest alone) to roughly 93% (rest + SRS + roster age) is the giveaway. This means that team strength is doing all the work, and rest is essentially riding along for the trip.

**This matches the intuition we put into the original hypothesis: load management is largely a luxury of strong organizations. Good teams have stars worth resting, and good teams win games. The two move together because they share a common cause, not because one drives the other.**

## What we found about the secondary question
The secondary question is harder: to see if the 2023 Player Participation Policy looks like it changed something, but we can't prove it with the data we have. While experiment 3 attempted to isolate star players from non-star top 3 minutes leaders, and compared rest rates before and after the policy took effect. The bar chart in “Broad Rest Rate by Group and Policy Period” shows the pattern cleanly: stars dropped from about 24% to 16%, while non-stars barely moved. The accompanying time series “Broad Rest Rate Over Time: Stars vs Non-Stars” shows star rest rates peaking sharply in 2023 (the last pre-policy season) and falling immediately afterward, while non-stars stay flat across the boundary.

That is exactly the visual signature a working policy would produce. A policy that targets stars should change star behavior and leave the comparison group alone, and that is what we see. The catch is that neither formal test crosses the conventional 0.05 significance threshold (the t-test on stars alone lands at p = 0.067), and the difference in differences interaction term comes in at p = 0.127. Both are borderline rather than null, and with only roughly 1.5 post-policy seasons in our data (more on this below) we are underpowered to push these into formal significance. 

So our answer to the secondary question is: **the policy appears to have meaningfully reduced classified rest events for the players it targets, but we cannot rule out that we are seeing relabeling rather than a true behavioral change, and a longer post-policy window is needed to settle it.**

## Connecting the two questions
These two questions are not really separate, and we want to make the link explicit because it is easy to miss. If rest does not independently buy wins (the primary finding), then it is fair to ask why teams rest stars in the first place. Part of the answer is player health and longevity, which our data cannot speak to. But part of it is also competitive ritual( load management became a default practice once Kawhi Leonard's 2018–19 championship run normalized it), and once a behavior becomes a norm in a league, every other team faces pressure to adopt it. 

That makes the secondary question matter even though the primary finding suggests rest does not pay off in regular-season wins. The league created the Player Participation Policy specifically because teams were resting stars in ways the league believed hurt the entertainment product, regardless of whether those rest decisions were actually helping the resting teams win games. Our two findings together suggest the policy is doing roughly what it was designed to do (curb the most visible cases of star rest) without disrupting a relationship between rest and winning that was already weak to nonexistent. The primary and secondary results are complementary pieces of the same picture: rest is widespread, rest does not buy wins on its own, and the league's intervention has dialed back rest for its intended targets without obvious collateral effects on team performance.


## The 2018-19 reporting issue
One result deserves its own paragraph because it kept showing up in different parts of the analysis and was easy to misread: the apparent rest-wins correlation in the full window is almost certainly a data artifact rather than a real basketball relationship.

Here is what happened. In 2017–18 and 2018–19, ProSportsTransactions notes were much less detailed than in later seasons — about 44% of relinquish events in 2017–18 and about 28% in 2018–19 had completely non-specific notes (just *"placed on IL"* with no further detail), as the “% Non-Specific Notes by Season” plot makes clear. Our keyword classifier reads those non-specific entries as zero rest events, because there is no soreness, tightness, or fatigue language to match on. So in those two early seasons, true rest behavior was systematically invisible to our proxy. Rest rates were mechanically pushed toward zero regardless of what teams actually did.

Now stack that against win percentage. In any given season some teams win a lot and others win very little, and our classifier is assigning artificially low rest rates to entire team-seasons in the 2018 and 2019 cohorts. The mathematics of correlation across the pooled 8-season dataset then picks up a spurious signal: the artificially-zero rest rates in those early seasons get correlated with the team-outcome variation in those same seasons, producing a small positive coefficient that has nothing to do with rest behavior. This is exactly why the correlation drops from r = 0.134 to r = 0.025 when we restrict the analysis to 2021–2025, where note quality is consistently high, and why it also vanishes in the full window once SRS is added as a control in Experiment 2. The "rest predicts winning" signal is not a real basketball effect; it is the residue of a data reporting convention that changed within our study window.

This is why we treat the restricted window as the primary sample throughout the analysis, and why the full-window numbers should be read as a cautionary tale about pooling data across reporting regimes rather than as substantive evidence either for or against rest.

## How this fits with the prior literature
The team-level null we land on is genuinely consistent with the small existing literature, which has always come at the question from the player side. Belk and colleagues (2017) compared NBA players who missed fewer than 5 rest games to those who missed 5–9, and found no meaningful playoff performance differences or injury-risk differences. Pradhan and Miller (2022) used a nested case control design through the 2020–21 season with Player Efficiency Rating and Win Shares as outcomes, and again found no evidence that strategic rest translated into better playoff performance. Our team-level analog of that question lands in the same place – once team quality is controlled, rest does not predict regular-season wins — which strengthens the broader case that the public reputation of load management may have outrun the empirical evidence.

The secondary question is where this project actually extends the literature. Bobby Marks's institutional reporting at ESPN walked through the policy's mechanics, but no published analysis we found has actually measured whether the policy changed team behavior in its target group. Our directional finding (stars down, non-stars flat) is the first empirical look at that question, and even though it is borderline rather than conclusive, it sets up a clear hypothesis for future seasons to test.


## What we still can't tell (limitations)
A few limitations matter much more than others for how the results should be read, and we want to be clear about which is which.

The two limitations that most affect interpretation:
The first is that **strategic rest is fundamentally unobservable**. We never see it directly , we only really infer it from free-text notes using a keyword classifier. The *Rest Detection Rate by Season: Narrow vs. Broad Definition* plot shows why this matters: the explicit-rest terms (*rest*, *load management*) are essentially never used, so we are forced to rely on the broad definition that also catches *soreness*, *tightness*, *fatigue*, and similar words. That broad bucket inevitably mixes genuine rest with real minor injuries, which means our rest_rate variable is best read as "rest plus minor-injury absences" rather than as a clean count of strategic rest. We could not have built a cleaner version of this variable from public data.

The second is the **post-2023 label substitution problem**, which is the exact failure mode our analysis cannot directly correct. After fines were introduced, teams have a financial incentive to call rest something more specific — *"left knee soreness"* or *"right hamstring tightness"* — that may evade even our broad classifier. This means the apparent drop in star rest after the policy could be a real behavioral change, or it could be teams getting more careful about the wording of their absence reports, or some mix of both. Our data alone cannot tell us which. This is the single biggest reason we describe the secondary finding as suggestive rather than conclusive.


### Additional Limitations
The third is that the 2019–20 and 2020–21 seasons were both shortened by COVID-19, with teams playing roughly 72 and 72 games respectively instead of the usual 82, on highly compressed schedules. That distorts season-level rest rates in both directions: fewer games means fewer opportunities for rest, but a compressed schedule means more back-to-backs and more incentive to rest. We did not separately adjust for this, and any direct comparison of absolute rest rates between those two seasons and the rest of the window should be read with that caveat in mind. The team-level “Broad Rest Rate by Team and Season” heatmap visibly shows different overall patterns in those COVID seasons, consistent with this concern.

The fourth is that the 2024–25 season in Dataset #2 is censored at January 12, 2025 because of the Kaggle snapshot date, so our post-policy window is effectively closer to 1.5 seasons rather than a full two. This is the most plausible reason our directional DiD result fails to cross significance. We are simply underpowered, not necessarily wrong.

## What's next
If this work continued past the class, the highest-value extensions would all sharpen the secondary analysis rather than reopen the primary one. The cleanest test would join the transaction log to the game schedule and flag nationally televised games (the only games where fines actually apply), turning the current league-wide DiD into a broadcast versus non-broadcast comparison that would also help separate real behavioral change from label substitution. Replacing the keyword classifier with a small hand-labeled training set would give us actual precision and recall numbers instead of relying on indirect sanity checks. A panel design with team fixed effects would absorb the persistent rest cultures visible in the heatmap (organizations like the Grizzlies and Spurs show long-standing philosophies that run across the entire window). And the simplest improvement of all would just be waiting: collecting full 2024–25 and 2025–26 seasons would likely resolve the borderline DiD significance without any methodological changes at all.


# Conclusion:
When team quality is held constant, rest does not predict winning. The apparent association is a confound, and the small full-window correlation that does appear is a data artifact driven by changes in note specificity in the early seasons rather than anything about basketball. The 2023 Player Participation Policy appears to have reduced reported rest for the star players it targets, but our data cannot separate a real behavioral change from label substitution, and the evidence is directional rather than conclusive given the short post-policy window. Both halves of our pre-registered hypothesis are supported at the level the data allows: rest reflects organizational quality rather than producing it, and the policy reduced reported star rest without disturbing that underlying relationship.
