## Research Question

Among NBA teams from the 2017–18 through 2024–25 regular seasons, is the rate of strategic rest games among a team's top-3 minutes leaders associated with regular season win percentage and playoff seeding, after controlling for team strength (preseason over/under win total), roster age, and back-to-back game frequency? As a secondary analysis, we explore whether the NBA's 2023 Player Participation Policy - which fines teams for resting "star" players (those named to All-Star or All-NBA teams in the prior three seasons) in nationally televised games — appears to have changed rest patterns for policy-eligible stars relative to non-star high-minutes players.


## Background and Prior Work

The practice of "load management" — strategically resting healthy NBA players during the regular season to preserve them for the playoffs — has become one of the most contentious topics in modern professional basketball. Once an obscure concept used by a small number of teams, load management entered mainstream attention when Kawhi Leonard sat out 22 of 82 regular season games during the Toronto Raptors' 2018-19 championship season, an approach widely credited with helping him win Finals MVP.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Since then, the practice has spread league-wide, generating concerns from broadcasters, fans, and the league office about competitive integrity and the value of regular-season games. The fundamental question for our project is whether this strategy is associated with team success, and whether the league's recent attempt to curb the practice has changed team behavior.

The empirical literature on regular-season rest and player outcomes is small but growing, and notably its findings have not strongly supported the assumed benefits of load management. Belk and colleagues conducted one of the earliest systematic analyses, examining 811 NBA players over the 2005-06 through 2014-15 seasons using DNP-CD ("did not play – coach's decision") records from Basketball Reference and Pro Sports Transactions.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Comparing players who missed fewer than 5 regular-season games due to rest against those who missed 5 to 9 games, they found no significant differences in playoff performance metrics or injury risk. The authors concluded that, contrary to widespread belief, regular-season rest did not improve playoff outcomes for the rested players themselves.

A more recent study by Pradhan and Miller extended this work using data from the 2016-17 through 2020-21 seasons, with rest decisions sourced from Spotrac.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Using Player Efficiency Rating and Win Shares as outcome measures, they similarly found insufficient evidence that strategic rest during the regular season translated into better post-season performance. A nested case-control comparison of 184 players (92 cases of rested players matched with 92 controls) showed no meaningful differences in performance changes from regular to post-season. Importantly, both studies focused on individual player performance rather than team-level outcomes such as win percentage or playoff seeding, and neither examined how the practice's prevalence has shifted over time. Our project addresses these gaps directly.

In response to growing concerns about load management's effect on regular-season competitiveness and broadcast value, the NBA Board of Governors in September 2023 approved the Player Participation Policy, which took effect at the start of the 2023-24 season.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) The policy formally defines a "star player" as any player named to an All-Star or All-NBA team in any of the prior three seasons, and prohibits teams from resting these players in nationally televised games or in-season tournament games without approved justification. Violations carry escalating fines beginning at USD 100,000 for a first offense and rising to USD 1.25 million for a third. Bobby Marks of ESPN provided detailed coverage of the policy's mechanics, including the specific rest restrictions, exception categories for older or high-mileage stars, and the league's investigative authority over suspect injury designations.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) This policy creates a natural setting for examining whether legal restrictions on rest decisions actually change team behavior — a question that has not yet been studied empirically given that the policy is recent.

Our project builds on the prior literature in three specific ways. First, we shift the unit of analysis from individual players to teams, asking whether a team's overall rest patterns are associated with regular-season win percentage and playoff seeding rather than individual player performance. Second, we extend the time window through the 2024-25 season, capturing both the pre- and post-policy eras. Third, by leveraging the policy's binary nature (it applies only to players meeting specific All-Star or All-NBA criteria), we can compare rest patterns for policy-eligible stars against those of non-star high-minutes players over the same period. This comparison group acts as a baseline to help separate the policy's effect from any general league-wide trends in rest behavior.

### Exploratory Data Analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import glob 
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# The cleaned dataset can be found at data/02-processed
csvFiles = sorted(glob.glob("data/02-processed/nba_*final*.csv"))

allData = []

for fileName in csvFiles:
    yearData = pd.read_csv(fileName)
    allData.append(yearData)

df = pd.concat(allData, ignore_index=True)
```

Analysis: Team Statistics Distributions
Win percentage and roster age both show consistent distributions across seasons, with win percentage ranging from approximately 0.17 to 0.83 across all eight seasons. The median win percentage hovers around 0.50, which is expected since half of all games have a winner and loser. Notably, roster age remains relatively stable across seasons (median ~26 years), with only modest year-to-year variation. This suggests that while individual teams may age their rosters, league-wide roster composition has remained fairly consistent from 2017-18 through 2024-25.

The stability in roster age is important for our research question: it indicates that any observed changes in team success or rest patterns are unlikely to be driven primarily by league-wide roster aging trends. This supports our decision to include roster age as a control variable—it captures team-level differences in composition without confounding secular trends over time. The variation in win percentage within each season (the spread of the violin plots) reflects genuine competitive differences between teams, which we explore further in subsequent analyses.

```python
print("Plot 1: Win percentage and roster age distributions by season.")
fig,axes = plt.subplots(1, 2, figsize=(13, 5))

sns.violinplot(data=df, x="season", y="win_pct", ax=axes[0])
axes[0].set_title("Win Percentage by Season")
axes[0].set_xlabel("Season")
axes[0].set_ylabel("Win Percentage")
axes[0].tick_params(axis="x", rotation=45)

sns.violinplot(data=df, x="season", y="roster_age", ax=axes[1])
axes[1].set_title("Roster Age by Season")
axes[1].set_xlabel("Season")
axes[1].set_ylabel("Average Roster Age")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

### Block 3: Roster Age and Playoff Outcomes

Roster age shows moderate positive correlation with both win percentage and playoff seeding, but the relationships are notably weaker than SRS. This suggests age reflects team composition rather than being a primary driver of success.

```python
# Plot 2: Correlation table for the main team statistics
correlationColumns = [
    "win_pct", "srs", "offensive_rating", "defensive_rating",
    "pace", "roster_age", "points_scored_per_game",
    "points_allowed_per_game", "playoff_seed"
]
correlationTable = df[correlationColumns].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlationTable, annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Correlation Heatmap of NBA Team Statistics")

plt.tight_layout()
plt.show()

srsWinCorrelation = correlationTable.loc["srs", "win_pct"]
paceWinCorrelation = correlationTable.loc["pace", "win_pct"]

print("This is heatmap shows the correlation between each variable used in this dataset and helps us validate (correlation vs causation also as mentioned in class).")
print(f"Analysis: SRS has a very strong relationship with win percentage (r = {srsWinCorrelation:.2f}), while pace has a much weaker relationship (r = {paceWinCorrelation:.2f}).")
```

### Block 2: Correlations Among Team Statistics

Correlation analysis reveals the strength of relationships between key variables. SRS shows the strongest correlation with win percentage (r = 0.96), followed by offensive rating (r = 0.68) and defensive rating (r = -0.58). Roster age correlates moderately with wins (r = 0.54) and playoff seeding (r = -0.52), while pace shows minimal correlation (r = -0.10).

```python
# Plot 3: Shows the main performance stats compared to win percentage.
fig,axes = plt.subplots(1, 3, figsize=(16, 5))

scatterPlots = [
    ("srs", "SRS vs Win Percentage"),
    ("offensive_rating", "Offensive Rating vs Win Percentage"),
    ("defensive_rating", "Defensive Rating vs Win Percentage")
]

for i, (xColumn, graphTitle) in enumerate(scatterPlots):
    sns.regplot(data=df, x=xColumn, y="win_pct", ax=axes[i], scatter_kws={"alpha": 0.6})
    rValue, pValue = stats.pearsonr(df[xColumn], df["win_pct"])

    axes[i].set_title(graphTitle + f"\nr = {rValue:.2f}")
    axes[i].set_xlabel(xColumn.replace("_", " ").title())
    axes[i].set_ylabel("Win Percentage")

plt.tight_layout()
plt.show()

srsR, srsP = stats.pearsonr(df["srs"], df["win_pct"])
offenseR, offenseP = stats.pearsonr(df["offensive_rating"], df["win_pct"])
defenseR, defenseP = stats.pearsonr(df["defensive_rating"], df["win_pct"])

print("Analysis: SRS is the strongest predictor of winning here (r = {srsR:.2f}); offense helps (r = {offenseR:.2f}), and better defense relates to more wins because defensive rating has a negative relationship (r = {defenseR:.2f}).")
```

### Analysis: Predictors of Team Success

The scatter plots confirm that SRS is the dominant predictor of win percentage (r = 0.96), with an almost perfect linear relationship. Teams rated as stronger than league average consistently win more games, validating our use of SRS as a control variable. Offensive rating shows a meaningful but weaker relationship (r = 0.68)—better offensive efficiency improves winning chances, but does not determine it alone. Defensive rating's negative correlation (r = -0.58) reflects that lower defensive rating (fewer points allowed per possession) associates with higher win percentage.

These relationships inform our understanding of load management's potential role. If strategic rest is associated with wins, we must ask: does rest cause wins, or do winning teams simply rest more because they can afford to? The strong SRS-wins relationship suggests the latter. Teams with higher preseason expectations (reflected in high SRS) both achieve more wins and likely have the organizational confidence to rest star players. Our control for team strength helps isolate any independent effect of rest patterns.

```python
# Plot 4: Here's the rosters age compared to win percentage and playoff seed.
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.regplot(data=df, x="roster_age", y="win_pct", ax=axes[0], scatter_kws={"alpha": 0.6})
ageWinR, ageWinP = stats.pearsonr(df["roster_age"], df["win_pct"])
axes[0].set_title(f"Roster Age vs Win Percentage\nr = {ageWinR:.2f}")
axes[0].set_xlabel("Roster Age")
axes[0].set_ylabel("Win Percentage")

sns.regplot(data=df, x="roster_age", y="playoff_seed", ax=axes[1], scatter_kws={"alpha": 0.6})
ageSeedR, ageSeedP = stats.pearsonr(df["roster_age"], df["playoff_seed"])
axes[1].set_title(f"Roster Age vs Playoff Seed\nr = {ageSeedR:.2f}")
axes[1].set_xlabel("Roster Age")
axes[1].set_ylabel("Playoff Seed")
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()

print("Analysis: Roster age is not strongly related to winning (r = {ageWinR:.2f}) or playoff seed (r = {ageSeedR:.2f}), so age alone does not explain team success very well.")
```

### Block 4: League-Wide Temporal Trends

From 2017-18 to 2024-25, median offensive rating increased from 108.8 to 115.0, and defensive rating increased from 109.0 to 115.1. Pace remained relatively stable (96.8–100.0 possessions per game), and roster age fluctuated without systematic trend. No sharp breaks appear at the 2023 policy boundary for any variable.

```python
# Plot 5: League-wide median trends over time.
trendColumns = ["offensive_rating", "defensive_rating", "pace", "roster_age"]
seasonTrendTable = df.groupby("season")[trendColumns].median().reset_index()

fig, axes = plt.subplots(2, 2, figsize=(13, 8))
axes = axes.flatten()

for i, columnName in enumerate(trendColumns):
    sns.lineplot(data=seasonTrendTable, x="season", y=columnName, marker="o", ax=axes[i])
    axes[i].axvline(2023.5, linestyle="--", color="red", alpha=0.7)
    axes[i].set_title("Average " + columnName.replace("_", " ").title() + " Over Time")
    axes[i].set_xlabel("Season")
    axes[i].set_ylabel("Median " + columnName.replace("_", " ").title())

plt.suptitle("League-Wide Median Trends Over Time (Red Line = After 2023 Policy)")
plt.tight_layout()
plt.show()

firstOffense = seasonTrendTable["offensive_rating"].iloc[0]
lastOffense = seasonTrendTable["offensive_rating"].iloc[-1]
firstPace = seasonTrendTable["pace"].iloc[0]
lastPace = seasonTrendTable["pace"].iloc[-1]

print(f"Analysis: Median offensive rating increased from {firstOffense:.1f} to {lastOffense:.1f}, while median pace changed less from {firstPace:.1f} to {lastPace:.1f}, showing scoring rose more than game speed.")
```

### Block 5: Dimensionality Reduction via PCA

PCA on seven standardized team statistics (SRS, offensive/defensive ratings, pace, roster age, points scored/allowed per game) yields two principal components explaining 77.4% of variation. PC1 (40.5%) represents overall team quality, with offensive rating, defensive rating, and SRS loading heavily. PC2 (36.9%) captures secondary patterns, likely pace-related.

PC1 correlates extremely strongly with win percentage (r = 0.94) and playoff seeding (r = -0.91), while PC2 shows weaker relationships with outcomes.

```python
# Plot 6: PCA combines multiple stats into two main dimensions.
pcaColumns = [
    "srs", "offensive_rating", "defensive_rating",
    "pace", "roster_age", "points_scored_per_game",
    "points_allowed_per_game"
]

pcaData = df[pcaColumns].dropna()

scaler = StandardScaler()
scaledData = scaler.fit_transform(pcaData)

pcaModel = PCA(n_components=2)
pcaScores = pcaModel.fit_transform(scaledData)

pcaTable = df.loc[pcaData.index, ["team", "season", "win_pct", "playoff_seed"]].copy()
pcaTable["PC1"] = pcaScores[:, 0]
pcaTable["PC2"] = pcaScores[:, 1]

loadingTable = pd.DataFrame(
    pcaModel.components_.T,
    columns=["PC1", "PC2"],
    index=pcaColumns
)

plt.figure(figsize=(9, 7))
sns.scatterplot(data=pcaTable, x="PC1", y="PC2", hue="win_pct", palette="viridis", s=70)

for statName in loadingTable.index:
    plt.arrow(0, 0, loadingTable.loc[statName, "PC1"] * 3, loadingTable.loc[statName, "PC2"] * 3,
              color="black", alpha=0.7, head_width=0.06)
    plt.text(loadingTable.loc[statName, "PC1"] * 3.2, loadingTable.loc[statName, "PC2"] * 3.2,
             statName, fontsize=9)

plt.title("PCA Biplot of NBA Team Statistics")
plt.xlabel(f"PC1 ({pcaModel.explained_variance_ratio_[0] * 100:.1f}% variance)")
plt.ylabel(f"PC2 ({pcaModel.explained_variance_ratio_[1] * 100:.1f}% variance)")
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)
plt.tight_layout()
plt.show()

pc1Variance = pcaModel.explained_variance_ratio_[0] * 100
pc2Variance = pcaModel.explained_variance_ratio_[1] * 100

print(f"Analysis: PCA reduces the team stats into two main patterns, with PC1 explaining {pc1Variance:.1f}% of the variation and PC2 explaining {pc2Variance:.1f}%.")
```

### Block 6: Principal Component Analysis (PCA)

PCA reduces multiple correlated team statistics into a smaller number of principal components, helping us identify the main patterns that distinguish successful from unsuccessful teams. By combining variables like offensive rating, defensive rating, pace, and roster age, PCA reveals which dimensions matter most for understanding team performance. PC1 and PC2 together explain over 77% of the variation in team statistics, suggesting that team success can be largely understood through two main patterns.

```python
# Plot 7: PC1 compared to actual team outcomes.
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.regplot(data=pcaTable, x="PC1", y="win_pct", ax=axes[0], scatter_kws={"alpha": 0.6})
pc1WinR, pc1WinP = stats.pearsonr(pcaTable["PC1"], pcaTable["win_pct"])
axes[0].set_title(f"PC1 vs Win Percentage\nr = {pc1WinR:.2f}")
axes[0].set_xlabel("PC1")
axes[0].set_ylabel("Win Percentage")

sns.regplot(data=pcaTable, x="PC1", y="playoff_seed", ax=axes[1], scatter_kws={"alpha": 0.6})
pc1SeedR, pc1SeedP = stats.pearsonr(pcaTable["PC1"], pcaTable["playoff_seed"])
axes[1].set_title(f"PC1 vs Playoff Seed\nr = {pc1SeedR:.2f}")
axes[1].set_xlabel("PC1")
axes[1].set_ylabel("Playoff Seed")
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()

print(f"Analysis: PC1 strongly connects to team success because it relates strongly to win percentage (r = {pc1WinR:.2f}) and playoff seed (r = {pc1SeedR:.2f}).")
```

## Analysis

### Analysis Part 1: What Drives Team Success?

Our EDA establishes that team success in the NBA is dominated by a single underlying quality dimension, captured by PC1 (which combines SRS, offensive/defensive efficiency). This dimension explains 94% of variation in wins and 91% of variation in playoff seeding. 

This finding has critical implications for our research question: **Any association between strategic rest and team success may partly reflect that strong teams (high PC1) both rest more and win more due to superior talent.** Our control variable strategy using SRS, offensive rating, and defensive rating as controls is therefore essential. Without controlling for team strength, we cannot isolate whether rest independently predicts success or merely correlates with it.

Roster age and pace, while interesting, show weak direct relationships with wins. This validates our choice to include them as controls without expecting them to be primary drivers.

### Analysis Part 2: Pre/Post Policy Baseline Changes

Between the pre-policy era (2017-18 to 2022-23) and post-policy era (2023-24 to 2024-25), offensive rating increased significantly (p = 0.000) and defensive rating increased significantly (p = 0.000). However, pace showed no significant change (p = 0.248), and roster age showed no significant change (p = 0.492).

These findings suggest:
1. **The 2023 policy did not fundamentally alter how teams construct rosters or play at different speeds.** If the policy drove behavioral changes in rest patterns, those changes occurred without requiring teams to shift roster composition or game pace.
2. **Offensive and defensive rating increases reflect broader league evolution** (three-point emphasis, rule changes, evolved tactics) rather than policy-driven changes in load management strategy.
3. **Any detected policy effect on rest must come from changes in rest decisions themselves, not from secondary changes in team construction.** This is methodologically cleaner for Dataset 2 analysis.

### Analysis Part 3: Team Quality as the Baseline Model

PC1 serves as a nearly perfect predictor of team success (r = 0.94 with win %), leaving limited residual variance for other factors to explain. When we eventually introduce rest patterns from Dataset 2, the critical question becomes: **Do rest decisions add explanatory power beyond team strength?**

Put differently:
- **If rest adds no value:** Teams that rest more simply happen to be stronger teams (confounding).
- **If rest adds value:** Better teams achieve better results partly because they strategically rest while maintaining competitiveness.
- **If rest subtracts value:** Strong teams could achieve even better results by resting less (load management is counterproductive).

Our baseline model team strength predicts wins is extremely strong. Rest patterns must overcome this strong baseline to show meaningful independent effects.
