### NBA Player Per-Game Statistics (2017-18 to 2024-25, Basketball Reference)

## Question 2a:
Dataset three is a player-level per-game statistics table scraped from Basketball Reference, with one CSV file per NBA regular season from 2017-18 through 2024-25. Each row represents a single player's per-game averages and season totals for one season. Across all eight files the dataset contains roughly 5,700 player-season rows, with each individual season containing between 650 and 815 rows. The sole purpose of this dataset in our project is to identify, for every team-season, the three players who led the team in total minutes played. Our research question is specifically about rest behavior among a team's top-3 minutes leaders, so we need this dataset to define who those players are before we can look up their absences in Dataset #2. The key variables we will use are:

- **Player:** Player name. Used as the join key to Dataset #2 after light name standardization for special characters and suffixes.

- **Player-additional:** Basketball Reference's internal player ID (e.g. embiijo01). A more reliable join key than the display name because it is unique and immune to spelling variants, accents, and name changes. We will keep this column rather than dropping it.

- **Team:** Team abbreviation for that row (e.g. LAL, BOS, GSW). Players traded mid-season have multiple rows, one per team they played for, plus an aggregate row. We will only use the per-team rows, not the aggregate rows, for identifying top-3 minutes leaders.

- **G (Games Played):** Number of regular-season games the player appeared in during that season. Combined with MP, this lets us compute total minutes.

- **MP (Minutes Per Game):** Average minutes played per game. This is a per-game average, not a season total. To get total minutes for the season we multiply MP by G. We rank players by total minutes (MP times G) rather than per-game average because a player who averaged 38 minutes in only 10 games should not outrank a player who played all 82 games at 30 minutes per game.

- **Age:** Player age during that season. Kept as a reference variable for any player-level context.

- **Awards:** A short concatenated string indicating any honors received that season, for example AS for All-Star or NBA1, NBA2, NBA3 for All-NBA First, Second, and Third Team. This column is critical for our secondary research question. The 2023 NBA Player Participation Policy defines a star as any player named to an All-Star or All-NBA team in any of the prior three seasons. We will parse this column to flag policy-eligible stars.

Derived variables we will construct from the raw fields:

- **Total_MP:** Total minutes for the season, computed as MP times G. This is the ranking variable used to identify each team's top-3 minutes leaders.

- **Top3_flag:** A boolean indicating whether the player is one of the three highest Total_MP players on their team for that season. This is the central output of this dataset and the link between Dataset #3 and Datasets #1 and #2.

- **Star_eligible:** A boolean flag for whether the player qualified as a star under the Player Participation Policy definition. Computed by checking whether the Awards column for that season or either of the two prior seasons contains AS, NBA1, NBA2, or NBA3.

- **Season:** Season label following the same convention as Datasets #1 and #2, the year the season ends, so 2023-24 is labeled 2024.

## Question 2b:
This dataset has two genuine concerns worth addressing and one structural quirk that needs to be handled in cleaning before the data is usable.

The first and most important issue is how Basketball Reference handles mid-season trades. When a player is traded during the season, the table contains multiple rows for that player, one per team they played for plus a combined aggregate row labeled TOT, 2TM, or 3TM depending on how many teams they appeared with. Across our eight seasons there are roughly 60 to 100 such aggregate rows per season, confirmed by the fact that every season file shows 32 or 33 distinct team values instead of the expected 30. For our project this is significant because the aggregate row gives the correct full-season totals but has no team attribution, while the per-team rows give the correct team attribution but only partial-season minutes. We will drop all aggregate rows and rank players within each team using only the per-team rows. A player traded mid-season will be evaluated as a top-3 candidate for each team they played on separately, using only the minutes they accumulated with that specific team.

The second concern is the Awards column, which we need for identifying policy-eligible stars. The column concatenates award codes without consistent separators, so a player who was an All-Star and All-NBA First Team member would appear as ASNBA1, and a player with multiple voting placements might appear as MVP-3CPOY-6ASNBA1. This means we cannot use simple word-boundary matching to extract the All-Star and All-NBA codes reliably. We will instead use a string contains approach, checking whether the Awards value contains AS, NBA1, NBA2, or NBA3 as recognizable substrings, and validate the resulting lists against the official NBA award announcements for each season to catch any misclassifications.

Finally, two non-issues worth naming so they are not mistaken for data quality problems. First, the Rk column is just a row counter from the original Basketball Reference web table and carries no meaning, we will drop it during cleaning. Second, exactly one row per season file has a missing MP value. This is the repeated column header row that Basketball Reference embeds periodically in the original HTML table. We will remove these rows by filtering out any row where the Player field equals the literal string Player.

```python
# =====================================================================
# 3A–3G: Load, Tidy, Audit, and Clean — NBA Per-Game Player Stats (Dataset #1)
# =====================================================================
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#  3A: LOAD ALL 8 SEASON CSVs 
RAW_DIR = 'data'

all_dfs = []
csv_files = sorted(glob.glob(os.path.join(RAW_DIR, 'nba_*_per_game.csv')))

print("=== 3A: RAW LOAD ===")
print(f"Found {len(csv_files)} season files:")

for fpath in csv_files:
    fname = os.path.basename(fpath)
    # Extract ending year from filename: nba_2017_18_per_game.csv → 2018
    parts = fname.split('_')          # ['nba', '2017', '18', 'per', 'game.csv']
    season_year = int(parts[2]) + 2000  # '18' → 2018, '19' → 2019, etc.

    tmp = pd.read_csv(fpath)
    tmp['season'] = season_year
    all_dfs.append(tmp)
    print(f"  {fname}  →  season={season_year}, shape={tmp.shape}")

df = pd.concat(all_dfs, ignore_index=True)
print(f"\nCombined raw shape: {df.shape}")

# 3B: TIDY 
print("\n=== 3B: TIDY ===")

for col in ['Player', 'Team', 'Pos']:
    df[col] = df[col].astype(str).str.strip().replace({'nan': np.nan, '': np.nan})

df['G']  = pd.to_numeric(df['G'],  errors='coerce')
df['MP'] = pd.to_numeric(df['MP'], errors='coerce')

n_before = len(df)
df = df[pd.to_numeric(df['Rk'], errors='coerce').notna()].reset_index(drop=True)
print(f"Dropped {n_before - len(df)} repeated header rows. Rows now: {len(df)}")

multi_team_players = df[df['Team'] == 'TOT']['Player'].unique()
df_single = df[~df['Player'].isin(multi_team_players)]
df_tot    = df[(df['Player'].isin(multi_team_players)) & (df['Team'] == 'TOT')]
df = pd.concat([df_single, df_tot], ignore_index=True)
print(f"After keeping TOT rows for traded players: {df.shape}")

df['total_mp'] = df['MP'] * df['G']

print("Tidy ✓  —  one row = one player-season")
print(df[['Player', 'Team', 'season', 'G', 'MP', 'total_mp']].head(8).to_string(index=False))

#  3C: SIZE AUDIT 
print("\n=== 3C: DATASET SIZE ===")
print(f"Total player-season rows : {len(df):,}")
print(f"Unique players           : {df['Player'].nunique():,}")
print(f"Unique teams             : {df['Team'].nunique()}")
print(f"Seasons covered          : {sorted(df['season'].unique())}")
print("\nRows per season:")
print(df['season'].value_counts().sort_index())

#  3D: MISSINGNESS AUDIT + CLEAN 
print("\n=== 3D: MISSING DATA AUDIT ===")
print("Missing values per column:")
print(df[['Player','Team','season','G','MP','total_mp']].isna().sum())

os.makedirs('data/01-interim', exist_ok=True)
plt.figure(figsize=(12, 5))
sns.heatmap(df[['Player','Team','G','MP','total_mp','season']].isna(),
            cbar=False, yticklabels=False, cmap='viridis')
plt.title('Missing data — NBA per-game stats (2017-18 through 2024-25)')
plt.tight_layout()
plt.savefig('data/01-interim/dataset1_missingness.png', dpi=150, bbox_inches='tight')
plt.show()

n_before = len(df)
df = df.dropna(subset=['Player', 'MP']).reset_index(drop=True)
print(f"Dropped {n_before - len(df)} rows missing Player or MP. Rows now: {len(df)}")

# 3E: OUTLIERS + SUSPICIOUS ENTRIES
print("\n=== 3E: OUTLIERS + SUSPICIOUS ENTRIES ===")

outlier_mp = df[df['MP'] > 48]
print(f"Rows with MP > 48 (impossible): {len(outlier_mp)}")
if len(outlier_mp) > 0:
    print(outlier_mp[['Player','Team','season','G','MP']].to_string(index=False))

low_games = df[df['G'] < 5]
print(f"\nPlayers with fewer than 5 games: {len(low_games)}  (noted — excluded from top-3 ranking below)")

print("\nMedian MP per game by season (sanity check — should be ~20-22 min):")
print(df.groupby('season')['MP'].median().round(1))

# ── 3F: BUILD TOP-3 MINUTES LEADERS LOOKUP TABLE 
print("\n=== 3F: TOP-3 MINUTES LEADERS LOOKUP TABLE ===")

df_qualified = df[df['G'] >= 20].copy()
print(f"Players with >= 20 games (qualified): {len(df_qualified):,}")

df_qualified['rank_mp'] = (
    df_qualified
    .groupby(['season', 'Team'])['total_mp']
    .rank(method='first', ascending=False)
)

top3 = (
    df_qualified[df_qualified['rank_mp'] <= 3]
    [['season', 'Team', 'Player', 'G', 'MP', 'total_mp', 'rank_mp']]
    .sort_values(['season', 'Team', 'rank_mp'])
    .reset_index(drop=True)
)

expected = 30 * 8 * 3   # 30 teams × 8 seasons × 3 players
print(f"\nTop-3 table shape: {top3.shape}  (expected ~{expected} rows)")

print("\nSample — 2024 Golden State Warriors:")
print(top3[(top3['Team'] == 'GSW') & (top3['season'] == 2024)].to_string(index=False))

print("\nSample — 2020 LA Lakers:")
print(top3[(top3['Team'] == 'LAL') & (top3['season'] == 2020)].to_string(index=False))

# ── 3G: SAVE 
print("\n=== 3G: SAVE ===")
os.makedirs('data/02-processed', exist_ok=True)

full_path = 'data/02-processed/nba_player_stats_clean.csv'
df.to_csv(full_path, index=False)
print(f"✓ Full cleaned stats table  →  {full_path}  {df.shape}")

top3_path = 'data/02-processed/nba_top3_leaders_by_team_season.csv'
top3.to_csv(top3_path, index=False)
print(f"✓ Top-3 lookup table        →  {top3_path}  {top3.shape}")

print("\n=== CLEANING JUSTIFICATION ===")
print(
    "1. Dropped repeated BBRef header rows (Rk = 'Rk') — these are formatting\n"
    "   artifacts from the HTML export, not real player rows.\n"
    "2. Multi-team (traded) players: kept only the TOT row to avoid double-\n"
    "   counting minutes across teams within the same season.\n"
    "3. Dropped rows missing Player name or MP — no useful data for ranking.\n"
    "4. Top-3 ranking requires >= 20 games played — excludes short-term signings\n"
    "   and injury returns who are not genuine high-usage starters.\n"
    "5. Ranked by total_mp (MP x G) not MP alone — correctly weights players\n"
    "   who played many games over those who played few games at high minutes."
)

print(f"\n✓ Final full table : {df.shape}")
print(f"✓ Final top-3 table: {top3.shape}")
```
