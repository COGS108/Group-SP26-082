Our control for team strength helps isolate any independent effect of rest patterns.

### Block 5: Dimensionality Reduction via PCA

PCA on seven standardized team statistics (SRS, offensive/defensive ratings, pace, roster age, points scored/allowed per game) yields two principal components explaining 77.4% of variation. PC1 (40.5%) represents overall team quality, with offensive rating, defensive rating, and SRS loading heavily. PC2 (36.9%) captures secondary patterns, likely pace-related.

PC1 correlates extremely strongly with win percentage (r = 0.94) and playoff seeding (r = -0.91), while PC2 shows weaker relationships with outcomes.

```python
# Plot 6: PCA combines multiple stats into two main dimensions.
pcaColumns = [
    "srs", "offensive_rating", "defensive_rating",
    "pace", "roster_age", "points_scored_per_game",
    "points_allowed_per_game"
]

pcaData = df[pcaColumns].dropna()

scaler = StandardScaler()
scaledData = scaler.fit_transform(pcaData)

pcaModel = PCA(n_components=2)
pcaScores = pcaModel.fit_transform(scaledData)

pcaTable = df.loc[pcaData.index, ["team", "season", "win_pct", "playoff_seed"]].copy()
pcaTable["PC1"] = pcaScores[:, 0]
pcaTable["PC2"] = pcaScores[:, 1]

loadingTable = pd.DataFrame(
    pcaModel.components_.T,
    columns=["PC1", "PC2"],
    index=pcaColumns
)

plt.figure(figsize=(9, 7))
sns.scatterplot(data=pcaTable, x="PC1", y="PC2", hue="win_pct", palette="viridis", s=70)

for statName in loadingTable.index:
    plt.arrow(0, 0, loadingTable.loc[statName, "PC1"] * 3, loadingTable.loc[statName, "PC2"] * 3,
              color="black", alpha=0.7, head_width=0.06)
    plt.text(loadingTable.loc[statName, "PC1"] * 3.2, loadingTable.loc[statName, "PC2"] * 3.2,
             statName, fontsize=9)

plt.title("PCA Biplot of NBA Team Statistics")
plt.xlabel(f"PC1 ({pcaModel.explained_variance_ratio_[0] * 100:.1f}% variance)")
plt.ylabel(f"PC2 ({pcaModel.explained_variance_ratio_[1] * 100:.1f}% variance)")
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)
plt.tight_layout()
plt.show()

pc1Variance = pcaModel.explained_variance_ratio_[0] * 100
pc2Variance = pcaModel.explained_variance_ratio_[1] * 100

print(f"Analysis: PCA reduces the team stats into two main patterns, with PC1 explaining {pc1Variance:.1f}% of the variation and PC2 explaining {pc2Variance:.1f}%.")
```

### Block 6: Principal Component Analysis (PCA)

PCA reduces multiple correlated team statistics into a smaller number of principal components, helping us identify the main patterns that distinguish successful from unsuccessful teams. By combining variables like offensive rating, defensive rating, pace, and roster age, PCA reveals which dimensions matter most for understanding team performance. PC1 and PC2 together explain over 77% of the variation in team statistics, suggesting that team success can be largely understood through two main patterns.

```python
# Plot 7: PC1 compared to actual team outcomes.
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.regplot(data=pcaTable, x="PC1", y="win_pct", ax=axes[0], scatter_kws={"alpha": 0.6})
pc1WinR, pc1WinP = stats.pearsonr(pcaTable["PC1"], pcaTable["win_pct"])
axes[0].set_title(f"PC1 vs Win Percentage\nr = {pc1WinR:.2f}")
axes[0].set_xlabel("PC1")
axes[0].set_ylabel("Win Percentage")

sns.regplot(data=pcaTable, x="PC1", y="playoff_seed", ax=axes[1], scatter_kws={"alpha": 0.6})
pc1SeedR, pc1SeedP = stats.pearsonr(pcaTable["PC1"], pcaTable["playoff_seed"])
axes[1].set_title(f"PC1 vs Playoff Seed\nr = {pc1SeedR:.2f}")
axes[1].set_xlabel("PC1")
axes[1].set_ylabel("Playoff Seed")
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()

print(f"Analysis: PC1 strongly connects to team success because it relates strongly to win percentage (r = {pc1WinR:.2f}) and playoff seed (r = {pc1SeedR:.2f}).")
```

## Analysis

### Analysis Part 1: What Drives Team Success?

Our EDA establishes that team success in the NBA is dominated by a single underlying quality dimension, captured by PC1 (which combines SRS, offensive/defensive efficiency). This dimension explains 94% of variation in wins and 91% of variation in playoff seeding. 

This finding has critical implications for our research question: **Any association between strategic rest and team success may partly reflect that strong teams (high PC1) both rest more and win more due to superior talent.** Our control variable strategy using SRS, offensive rating, and defensive rating as controls is therefore essential. Without controlling for team strength, we cannot isolate whether rest independently predicts success or merely correlates with it.

Roster age and pace, while interesting, show weak direct relationships with wins. This validates our choice to include them as controls without expecting them to be primary drivers.

### Analysis Part 2: Pre/Post Policy Baseline Changes

Between the pre-policy era (2017-18 to 2022-23) and post-policy era (2023-24 to 2024-25), offensive rating increased significantly (p = 0.000) and defensive rating increased significantly (p = 0.000). However, pace showed no significant change (p = 0.248), and roster age showed no significant change (p = 0.492).

These findings suggest:
1. **The 2023 policy did not fundamentally alter how teams construct rosters or play at different speeds.** If the policy drove behavioral changes in rest patterns, those changes occurred without requiring teams to shift roster composition or game pace.
2. **Offensive and defensive rating increases reflect broader league evolution** (three-point emphasis, rule changes, evolved tactics) rather than policy-driven changes in load management strategy.
3. **Any detected policy effect on rest must come from changes in rest decisions themselves, not from secondary changes in team construction.** This is methodologically cleaner for Dataset 2 analysis.

### Analysis Part 3: Team Quality as the Baseline Model

PC1 serves as a nearly perfect predictor of team success (r = 0.94 with win %), leaving limited residual variance for other factors to explain. When we eventually introduce rest patterns from Dataset 2, the critical question becomes: **Do rest decisions add explanatory power beyond team strength?**

Put differently:
- **If rest adds no value:** Teams that rest more simply happen to be stronger teams (confounding).
- **If rest adds value:** Better teams achieve better results partly because they strategically rest while maintaining competitiveness.
- **If rest subtracts value:** Strong teams could achieve even better results by resting less (load management is counterproductive).

Our baseline model team strength predicts wins is extremely strong. Rest patterns must overcome this strong baseline to show meaningful independent effects.
