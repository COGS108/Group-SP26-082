### NBA Player Per-Game Statistics (2017-18 to 2024-25, Basketball Reference)

## Question 2a:
Dataset three is a player-level per-game statistics table scraped from Basketball 
Reference, with one CSV file per NBA regular season from 2017-18 through 2024-25. 
Each row in the cleaned dataset represents a single player's statistics for one 
team during one season. After combining all 8 files we start with 5,697 raw rows. 
We drop 8 repeated header rows (a Basketball Reference HTML formatting artifact) 
and 610 aggregate rows created for players who were traded mid-season, leaving 
5,079 clean player-team-season rows covering 1,291 unique players across all 30 
NBA teams. The purpose of this dataset in our project is specifically to identify, 
for every team-season, the three players who led that team in total minutes played. 
Those three players are the subjects of our load management analysis. The key 
variables we use are:

- **Player:** Player name, used as the join key to Dataset #2 after light name 
standardization for special characters and suffixes.

- **Team:** The team abbreviation for that specific row (e.g. LAL, BOS, GSW). 
For traded players we keep one row per team they played for, so their minutes 
are attributed to each specific team rather than combined into a season total.

- **G (Games Played):** Number of regular-season games the player appeared in 
for that specific team. A player traded mid-season will have a separate G value 
for each team they played for. We require a minimum of 20 games for a player to 
qualify for a team's top-3 ranking, since anything below that is a short-term 
stint rather than a genuine rotation role.

- **MP (Minutes Per Game):** Average minutes played per game for that team. 
This is a per-game average, not a season total. To get total minutes we multiply 
MP by G.

- **total_mp (derived):** Total minutes played for that team, computed as MP 
times G. This is the variable we rank players by within each team-season to 
identify the top-3 minutes leaders. We use total minutes rather than per-game 
average because a player who averaged 38 minutes in 10 games should not outrank 
a player who played all 70 games at 30 minutes per game.

- **Awards:** A concatenated string of any honors the player received that season, 
for example AS for All-Star and NBA1, NBA2, NBA3 for All-NBA First, Second, and 
Third Team. We parse this column to flag players who are eligible under the 2023 
NBA Player Participation Policy, which defines a star as any player who received 
one of these honors in any of the three prior seasons.

The final top-3 lookup table contains exactly 720 rows, 3 players for each of 
the 30 teams across each of the 8 seasons, with all 30 teams represented in 
every season.

## Question 2b:
This dataset has two genuine concerns worth addressing and one structural quirk that needs to be handled in cleaning before the data is usable.

The first and most important issue is how Basketball Reference handles mid-season trades. When a player is traded during the season, the table contains multiple rows for that player, one per team they played for plus a combined aggregate row labeled TOT, 2TM, or 3TM depending on how many teams they appeared with. Across our eight seasons there are roughly 60 to 100 such aggregate rows per season, confirmed by the fact that every season file shows 32 or 33 distinct team values instead of the expected 30. For our project this is significant because the aggregate row gives the correct full-season totals but has no team attribution, while the per-team rows give the correct team attribution but only partial-season minutes. We will drop all aggregate rows and rank players within each team using only the per-team rows. A player traded mid-season will be evaluated as a top-3 candidate for each team they played on separately, using only the minutes they accumulated with that specific team.

The second concern is the Awards column, which we need for identifying policy-eligible stars. The column concatenates award codes without consistent separators, so a player who was an All-Star and All-NBA First Team member would appear as ASNBA1, and a player with multiple voting placements might appear as MVP-3CPOY-6ASNBA1. This means we cannot use simple word-boundary matching to extract the All-Star and All-NBA codes reliably. We will instead use a string contains approach, checking whether the Awards value contains AS, NBA1, NBA2, or NBA3 as recognizable substrings, and validate the resulting lists against the official NBA award announcements for each season to catch any misclassifications.

Finally, two non-issues worth naming so they are not mistaken for data quality problems. First, the Rk column is just a row counter from the original Basketball Reference web table and carries no meaning, we will drop it during cleaning. Second, exactly one row per season file has a missing MP value. This is the repeated column header row that Basketball Reference embeds periodically in the original HTML table. We will remove these rows by filtering out any row where the Player field equals the literal string Player.

```python
# =====================================================================
# 3A-3G: Load, Tidy, Audit, and Clean — NBA Per-Game Player Stats (Dataset #3)
# =====================================================================
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- 3A: LOAD ALL 8 SEASON CSVs ---------------------------------------
RAW_DIR = 'data/00-raw'

all_dfs = []
csv_files = sorted(glob.glob(os.path.join(RAW_DIR, 'nba_*_per_game.csv')))

print("=== 3A: RAW LOAD ===")
print(f"Found {len(csv_files)} season files:")

for fpath in csv_files:
    fname = os.path.basename(fpath)
    parts = fname.split('_')
    season_year = int(parts[2]) + 2000
    tmp = pd.read_csv(fpath)
    tmp['season'] = season_year
    all_dfs.append(tmp)
    print(f"  {fname}  ->  season={season_year}, shape={tmp.shape}")

df = pd.concat(all_dfs, ignore_index=True)
print(f"\nCombined raw shape: {df.shape}")

# --- 3B: TIDY ---------------------------------------------------------
print("\n=== 3B: TIDY ===")

for col in ['Player', 'Team', 'Pos']:
    df[col] = df[col].astype(str).str.strip().replace({'nan': np.nan, '': np.nan})

df['G']  = pd.to_numeric(df['G'],  errors='coerce')
df['MP'] = pd.to_numeric(df['MP'], errors='coerce')

# Drop repeated BBRef header rows
n_before = len(df)
df = df[pd.to_numeric(df['Rk'], errors='coerce').notna()].reset_index(drop=True)
print(f"Dropped {n_before - len(df)} repeated header rows. Rows now: {len(df)}")

# Drop aggregate rows for traded players — keep per-team rows only
# Each player's minutes are attributed to the specific team they played for
n_before = len(df)
df = df[~df['Team'].isin(['TOT', '2TM', '3TM', '4TM'])].reset_index(drop=True)
print(f"Dropped {n_before - len(df)} aggregate rows for traded players. Rows now: {len(df)}")

# Compute total minutes: MP (per game average) x G (games played for that team)
df['total_mp'] = df['MP'] * df['G']

print("Tidy check: one row = one player-team-season")
print(df[['Player', 'Team', 'season', 'G', 'MP', 'total_mp']].head(8).to_string(index=False))

# --- 3C: SIZE AUDIT ---------------------------------------------------
print("\n=== 3C: DATASET SIZE ===")
print(f"Total player-team-season rows : {len(df):,}")
print(f"Unique players                : {df['Player'].nunique():,}")
print(f"Unique teams                  : {df['Team'].nunique()}")
print(f"Seasons covered               : {sorted(df['season'].unique())}")

multi = df.groupby(['Player', 'season'])['Team'].nunique()
print(f"\nPlayer-seasons split across 2+ teams (traded): {(multi > 1).sum()}")

print("\nRows per season:")
print(df['season'].value_counts().sort_index())

# --- 3D: MISSINGNESS AUDIT --------------------------------------------
print("\n=== 3D: MISSING DATA AUDIT ===")
print("Missing values per column:")
print(df[['Player', 'Team', 'season', 'G', 'MP', 'total_mp']].isna().sum())

os.makedirs('data/01-interim', exist_ok=True)
plt.figure(figsize=(12, 5))
sns.heatmap(df[['Player', 'Team', 'G', 'MP', 'total_mp', 'season']].isna(),
            cbar=False, yticklabels=False, cmap='viridis')
plt.title('Missing data -- NBA per-game stats (2017-18 through 2024-25)')
plt.tight_layout()
plt.savefig('data/01-interim/dataset3_missingness.png', dpi=150, bbox_inches='tight')
plt.show()

n_before = len(df)
df = df.dropna(subset=['Player', 'MP']).reset_index(drop=True)
print(f"Dropped {n_before - len(df)} rows missing Player or MP. Rows now: {len(df)}")

# --- 3E: OUTLIERS + SUSPICIOUS ENTRIES --------------------------------
print("\n=== 3E: OUTLIERS + SUSPICIOUS ENTRIES ===")

outlier_mp = df[df['MP'] > 48]
print(f"Rows with MP > 48 (impossible in regulation): {len(outlier_mp)}")
if len(outlier_mp) > 0:
    print(outlier_mp[['Player', 'Team', 'season', 'G', 'MP']].to_string(index=False))

low_games = df[df['G'] < 20]
print(f"\nPlayer-team stints with fewer than 20 games: {len(low_games)}")
print("  (excluded from top-3 ranking -- not genuine rotation players for that team)")

print("\nMedian MP per game by season:")
print(df.groupby('season')['MP'].median().round(1))

# --- 3F: TOP-3 MINUTES LEADERS LOOKUP TABLE --------------------------
print("\n=== 3F: TOP-3 MINUTES LEADERS LOOKUP TABLE ===")

# Require >= 20 games for that specific team stint to qualify
df_qualified = df[df['G'] >= 20].copy()
print(f"Player-team stints with >= 20 games (qualified): {len(df_qualified):,}")

df_qualified['rank_mp'] = (
    df_qualified
    .groupby(['season', 'Team'])['total_mp']
    .rank(method='first', ascending=False)
)

top3 = (
    df_qualified[df_qualified['rank_mp'] <= 3]
    [['season', 'Team', 'Player', 'G', 'MP', 'total_mp', 'rank_mp']]
    .sort_values(['season', 'Team', 'rank_mp'])
    .reset_index(drop=True)
)

expected = 30 * 8 * 3
print(f"\nTop-3 table shape: {top3.shape}  (expected {expected} rows)")
print(f"All 30 teams present every season: {(top3.groupby('season')['Team'].nunique() == 30).all()}")

print("\nSample -- 2024 Golden State Warriors:")
print(top3[(top3['Team'] == 'GSW') & (top3['season'] == 2024)].to_string(index=False))

print("\nSample -- 2020 LA Lakers:")
print(top3[(top3['Team'] == 'LAL') & (top3['season'] == 2020)].to_string(index=False))

# --- 3G: SAVE ---------------------------------------------------------
print("\n=== 3G: SAVE ===")
os.makedirs('data/02-processed', exist_ok=True)

full_path = 'data/02-processed/nba_player_stats_clean.csv'
df.to_csv(full_path, index=False)
print(f"Full cleaned stats table  ->  {full_path}  {df.shape}")

top3_path = 'data/02-processed/nba_top3_leaders_by_team_season.csv'
top3.to_csv(top3_path, index=False)
print(f"Top-3 lookup table        ->  {top3_path}  {top3.shape}")

print("\n=== CLEANING JUSTIFICATION ===")
print(
    "1. Dropped repeated BBRef header rows (Rk = 'Rk') -- formatting artifacts\n"
    "   from the HTML export, not real player rows.\n"
    "2. Dropped aggregate rows (TOT, 2TM, 3TM, 4TM) and kept per-team rows for\n"
    "   traded players. Each player's minutes are attributed to the specific team\n"
    "   they played for, so a traded player can qualify for the top-3 of each team\n"
    "   they contributed to based on their games played for that team specifically.\n"
    "3. Dropped rows missing Player name or MP -- no useful data for ranking.\n"
    "4. Top-3 ranking requires >= 20 games for that specific team stint -- excludes\n"
    "   short-term signings and 10-day contracts who are not genuine rotation players.\n"
    "5. Ranked by total_mp (MP x G), not per-game MP alone -- correctly weights\n"
    "   players who played many games over those with few high-minute appearances."
)

print(f"\nFinal full table : {df.shape}")
print(f"Final top-3 table: {top3.shape}")
```
