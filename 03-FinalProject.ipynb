```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import glob 
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# The cleaned dataset can be found at data/02-processed
csvFiles = sorted(glob.glob("data/02-processed/nba_*final*.csv"))

allData = []

for fileName in csvFiles:
    yearData = pd.read_csv(fileName)
    allData.append(yearData)

df = pd.concat(allData, ignore_index=True)

#Tester to see if its working (Ignore)
print("Rows and columns:", df.shape)
print("Seasons:", sorted(df["season"].unique()))
print(df.head())
```

Analysis: Team Statistics Distributions
Win percentage and roster age both show consistent distributions across seasons, with win percentage ranging from approximately 0.17 to 0.83 across all eight seasons. The median win percentage hovers around 0.50, which is expected since half of all games have a winner and loser. Notably, roster age remains relatively stable across seasons (median ~26 years), with only modest year-to-year variation. This suggests that while individual teams may age their rosters, league-wide roster composition has remained fairly consistent from 2017-18 through 2024-25.

The stability in roster age is important for our research question: it indicates that any observed changes in team success or rest patterns are unlikely to be driven primarily by league-wide roster aging trends. This supports our decision to include roster age as a control variable—it captures team-level differences in composition without confounding secular trends over time. The variation in win percentage within each season (the spread of the violin plots) reflects genuine competitive differences between teams, which we explore further in subsequent analyses.

```python
print("Plot 1: Win percentage and roster age distributions by season.")
fig,axes = plt.subplots(1, 2, figsize=(13, 5))

sns.violinplot(data=df, x="season", y="win_pct", ax=axes[0])
axes[0].set_title("Win Percentage by Season")
axes[0].set_xlabel("Season")
axes[0].set_ylabel("Win Percentage")
axes[0].tick_params(axis="x", rotation=45)

sns.violinplot(data=df, x="season", y="roster_age", ax=axes[1])
axes[1].set_title("Roster Age by Season")
axes[1].set_xlabel("Season")
axes[1].set_ylabel("Average Roster Age")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

Analysis: Team Statistics Distributions

```python
# Plot 2: Correlation table for the main team statistics
correlationColumns = [
    "win_pct", "srs", "offensive_rating", "defensive_rating",
    "pace", "roster_age", "points_scored_per_game",
    "points_allowed_per_game", "playoff_seed"
]
correlationTable = df[correlationColumns].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlationTable, annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Correlation Heatmap of NBA Team Statistics")

plt.tight_layout()
plt.show()

srsWinCorrelation = correlationTable.loc["srs", "win_pct"]
paceWinCorrelation = correlationTable.loc["pace", "win_pct"]

print("This is heatmap shows the correlation between each variable used in this dataset and helps us validate (correlation vs causation also as mentioned in class).")
print(f"Analysis: SRS has a very strong relationship with win percentage (r = {srsWinCorrelation:.2f}), while pace has a much weaker relationship (r = {paceWinCorrelation:.2f}).")
```

Analysis: Correlation with Team Success
The correlation heatmap reveals that Simple Rating System (SRS) is the strongest predictor of win percentage (r = 0.96), indicating that teams rated as stronger relative to league average almost universally achieve higher win percentages. This strong relationship validates SRS as our primary control variable for team strength. Offensive rating also shows a moderate positive correlation with wins (r = 0.68), while defensive rating shows a moderate negative correlation (r = -0.58)—meaning better defense (lower defensive rating) associates with more wins, as expected.

Importantly, pace (possessions per game) shows a weak relationship with win percentage (r = -0.10), suggesting that playing faster or slower does not inherently determine team success when controlling for other factors. Roster age shows a moderate positive correlation with wins (r = 0.54), but this may reflect confounding: older rosters often belong to established, successful franchises rather than age itself improving performance.

These correlations set the foundation for understanding our research question. Since SRS strongly predicts wins, any association between strategic rest and wins may partly reflect that strong teams (high SRS) both win more and can afford to rest players. We address this by controlling for team strength in our analysis.

```python
# Plot 3: Shows the main performance stats compared to win percentage.
fig,axes = plt.subplots(1, 3, figsize=(16, 5))

scatterPlots = [
    ("srs", "SRS vs Win Percentage"),
    ("offensive_rating", "Offensive Rating vs Win Percentage"),
    ("defensive_rating", "Defensive Rating vs Win Percentage")
]

for i, (xColumn, graphTitle) in enumerate(scatterPlots):
    sns.regplot(data=df, x=xColumn, y="win_pct", ax=axes[i], scatter_kws={"alpha": 0.6})
    rValue, pValue = stats.pearsonr(df[xColumn], df["win_pct"])

    axes[i].set_title(graphTitle + f"\nr = {rValue:.2f}")
    axes[i].set_xlabel(xColumn.replace("_", " ").title())
    axes[i].set_ylabel("Win Percentage")

plt.tight_layout()
plt.show()

srsR, srsP = stats.pearsonr(df["srs"], df["win_pct"])
offenseR, offenseP = stats.pearsonr(df["offensive_rating"], df["win_pct"])
defenseR, defenseP = stats.pearsonr(df["defensive_rating"], df["win_pct"])

print("Analysis: SRS is the strongest predictor of winning here (r = {srsR:.2f}); offense helps (r = {offenseR:.2f}), and better defense relates to more wins because defensive rating has a negative relationship (r = {defenseR:.2f}).")
```

### Analysis: Predictors of Team Success

The scatter plots confirm that SRS is the dominant predictor of win percentage (r = 0.96), with an almost perfect linear relationship. Teams rated as stronger than league average consistently win more games, validating our use of SRS as a control variable. Offensive rating shows a meaningful but weaker relationship (r = 0.68)—better offensive efficiency improves winning chances, but does not determine it alone. Defensive rating's negative correlation (r = -0.58) reflects that lower defensive rating (fewer points allowed per possession) associates with higher win percentage.

These relationships inform our understanding of load management's potential role. If strategic rest is associated with wins, we must ask: does rest cause wins, or do winning teams simply rest more because they can afford to? The strong SRS-wins relationship suggests the latter. Teams with higher preseason expectations (reflected in high SRS) both achieve more wins and likely have the organizational confidence to rest star players. Our control for team strength helps isolate any independent effect of rest patterns.

```python
# Plot 4: Here's the rosters age compared to win percentage and playoff seed.
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.regplot(data=df, x="roster_age", y="win_pct", ax=axes[0], scatter_kws={"alpha": 0.6})
ageWinR, ageWinP = stats.pearsonr(df["roster_age"], df["win_pct"])
axes[0].set_title(f"Roster Age vs Win Percentage\nr = {ageWinR:.2f}")
axes[0].set_xlabel("Roster Age")
axes[0].set_ylabel("Win Percentage")

sns.regplot(data=df, x="roster_age", y="playoff_seed", ax=axes[1], scatter_kws={"alpha": 0.6})
ageSeedR, ageSeedP = stats.pearsonr(df["roster_age"], df["playoff_seed"])
axes[1].set_title(f"Roster Age vs Playoff Seed\nr = {ageSeedR:.2f}")
axes[1].set_xlabel("Roster Age")
axes[1].set_ylabel("Playoff Seed")
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()

print("Analysis: Roster age is not strongly related to winning (r = {ageWinR:.2f}) or playoff seed (r = {ageSeedR:.2f}), so age alone does not explain team success very well.")
```

### Analysis: Roster Age as a Control Variable

Roster age shows moderate positive correlations with both win percentage (r = 0.54) and playoff seeding (r = -0.52), but these relationships are substantially weaker than SRS. This indicates that while older rosters may have slight advantages, age alone does not strongly predict team success. The weak relationship suggests that roster age primarily functions as a confounding variable—older rosters often belong to established, veteran-led organizations that win for other reasons (better talent, organizational stability, coaching).
By including roster age as a control variable in our analysis, we account for this confounding while recognizing its limited direct predictive power. Teams with very young rosters (e.g., rebuilding teams) and very old rosters (e.g., veteran contenders) can both achieve high win percentages if their talent levels differ. This reinforces our focus on team strength (SRS) and rest patterns as the primary drivers of success, with age serving as an important but secondary control.

```python
# Plot 5: League-wide median trends over time.
trendColumns = ["offensive_rating", "defensive_rating", "pace", "roster_age"]
seasonTrendTable = df.groupby("season")[trendColumns].median().reset_index()

fig, axes = plt.subplots(2, 2, figsize=(13, 8))
axes = axes.flatten()

for i, columnName in enumerate(trendColumns):
    sns.lineplot(data=seasonTrendTable, x="season", y=columnName, marker="o", ax=axes[i])
    axes[i].axvline(2023.5, linestyle="--", color="red", alpha=0.7)
    axes[i].set_title("Average " + columnName.replace("_", " ").title() + " Over Time")
    axes[i].set_xlabel("Season")
    axes[i].set_ylabel("Median " + columnName.replace("_", " ").title())

plt.suptitle("League-Wide Median Trends Over Time (Red Line = After 2023 Policy)")
plt.tight_layout()
plt.show()

firstOffense = seasonTrendTable["offensive_rating"].iloc[0]
lastOffense = seasonTrendTable["offensive_rating"].iloc[-1]
firstPace = seasonTrendTable["pace"].iloc[0]
lastPace = seasonTrendTable["pace"].iloc[-1]

print(f"Analysis: Median offensive rating increased from {firstOffense:.1f} to {lastOffense:.1f}, while median pace changed less from {firstPace:.1f} to {lastPace:.1f}, showing scoring rose more than game speed.")
```

he absence of dramatic breaks at the 2023 policy line for pace and roster age indicates that the policy did not fundamentally alter how teams construct rosters or pace their games.
