## Research Question

Among NBA teams from the 2017–18 through 2024–25 regular seasons, is the rate of strategic rest games among a team's top-3 minutes leaders associated with regular season win percentage and playoff seeding, after controlling for team strength (preseason over/under win total), roster age, and back-to-back game frequency? ## Background and Prior Work

The practice of "load management" — strategically resting healthy NBA players during the regular season to preserve them for the playoffs — has become one of the most contentious topics in modern professional basketball. Once an obscure concept used by a small number of teams, load management entered mainstream attention when Kawhi Leonard sat out 22 of 82 regular season games during the Toronto Raptors' 2018-19 championship season, an approach widely credited with helping him win Finals MVP.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Since then, the practice has spread league-wide, generating concerns from broadcasters, fans, and the league office about competitive integrity and the value of regular-season games. The fundamental question for our project is whether this strategy is associated with team success, and whether the league's recent attempt to curb the practice has changed team behavior.

The empirical literature on regular-season rest and player outcomes is small but growing, and notably its findings have not strongly supported the assumed benefits of load management. Belk and colleagues conducted one of the earliest systematic analyses, examining 811 NBA players over the 2005-06 through 2014-15 seasons using DNP-CD ("did not play – coach's decision") records from Basketball Reference and Pro Sports Transactions.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Comparing players who missed fewer than 5 regular-season games due to rest against those who missed 5 to 9 games, they found no significant differences in playoff performance metrics or injury risk. The authors concluded that, contrary to widespread belief, regular-season rest did not improve playoff outcomes for the rested players themselves.

A more recent study by Pradhan and Miller extended this work using data from the 2016-17 through 2020-21 seasons, with rest decisions sourced from Spotrac.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Using Player Efficiency Rating and Win Shares as outcome measures, they similarly found insufficient evidence that strategic rest during the regular season translated into better post-season performance. A nested case-control comparison of 184 players (92 cases of rested players matched with 92 controls) showed no meaningful differences in performance changes from regular to post-season. Importantly, both studies focused on individual player performance rather than team-level outcomes such as win percentage or playoff seeding, and neither examined how the practice's prevalence has shifted over time. Our project addresses these gaps directly.

In response to growing concerns about load management's effect on regular-season competitiveness and broadcast value, the NBA Board of Governors in September 2023 approved the Player Participation Policy, which took effect at the start of the 2023-24 season.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) The policy formally defines a "star player" as any player named to an All-Star or All-NBA team in any of the prior three seasons, and prohibits teams from resting these players in nationally televised games or in-season tournament games without approved justification. Violations carry escalating fines beginning at USD 100,000 for a first offense and rising to USD 1.25 million for a third. Bobby Marks of ESPN provided detailed coverage of the policy's mechanics, including the specific rest restrictions, exception categories for older or high-mileage stars, and the league's investigative authority over suspect injury designations.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) This policy creates a natural setting for examining whether legal restrictions on rest decisions actually change team behavior — a question that has not yet been studied empirically given that the policy is recent.

Our project builds on the prior literature in three specific ways. First, we shift the unit of analysis from individual players to teams, asking whether a team's overall rest patterns are associated with regular-season win percentage and playoff seeding rather than individual player performance. Second, we extend the time window through the 2024-25 season, capturing both the pre- and post-policy eras. Third, by leveraging the policy's binary nature (it applies only to players meeting specific All-Star or All-NBA criteria), we can compare rest patterns for policy-eligible stars against those of non-star high-minutes players over the same period. This comparison group acts as a baseline to help separate the policy's effect from any general league-wide trends in rest behavior.

```python
# =====================================================================
# 3A-3G: Load, Tidy, Audit, and Clean — NBA Per-Game Player Stats (Dataset #3)
# =====================================================================
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- 3A: LOAD ALL 8 SEASON CSVs ---------------------------------------
RAW_DIR = 'data'

all_dfs = []
csv_files = sorted(glob.glob(os.path.join(RAW_DIR, 'nba_*_per_game.csv')))

print("=== 3A: RAW LOAD ===")
print(f"Found {len(csv_files)} season files:")

for fpath in csv_files:
    fname = os.path.basename(fpath)
    parts = fname.split('_')
    season_year = int(parts[2]) + 2000
    tmp = pd.read_csv(fpath)
    tmp['season'] = season_year
    all_dfs.append(tmp)
    print(f"  {fname}  ->  season={season_year}, shape={tmp.shape}")

df = pd.concat(all_dfs, ignore_index=True)
print(f"\nCombined raw shape: {df.shape}")

# --- 3B: TIDY ---------------------------------------------------------
print("\n=== 3B: TIDY ===")

for col in ['Player', 'Team', 'Pos']:
    df[col] = df[col].astype(str).str.strip().replace({'nan': np.nan, '': np.nan})

df['G']  = pd.to_numeric(df['G'],  errors='coerce')
df['MP'] = pd.to_numeric(df['MP'], errors='coerce')

# Drop repeated BBRef header rows
n_before = len(df)
df = df[pd.to_numeric(df['Rk'], errors='coerce').notna()].reset_index(drop=True)
print(f"Dropped {n_before - len(df)} repeated header rows. Rows now: {len(df)}")

# Drop aggregate rows for traded players — keep per-team rows only
# Each player's minutes are attributed to the specific team they played for
n_before = len(df)
df = df[~df['Team'].isin(['TOT', '2TM', '3TM', '4TM'])].reset_index(drop=True)
print(f"Dropped {n_before - len(df)} aggregate rows for traded players. Rows now: {len(df)}")

# Compute total minutes: MP (per game average) x G (games played for that team)
df['total_mp'] = df['MP'] * df['G']

print("Tidy check: one row = one player-team-season")
print(df[['Player', 'Team', 'season', 'G', 'MP', 'total_mp']].head(8).to_string(index=False))

# --- 3C: SIZE AUDIT ---------------------------------------------------
print("\n=== 3C: DATASET SIZE ===")
print(f"Total player-team-season rows : {len(df):,}")
print(f"Unique players                : {df['Player'].nunique():,}")
print(f"Unique teams                  : {df['Team'].nunique()}")
print(f"Seasons covered               : {sorted(df['season'].unique())}")

multi = df.groupby(['Player', 'season'])['Team'].nunique()
print(f"\nPlayer-seasons split across 2+ teams (traded): {(multi > 1).sum()}")

print("\nRows per season:")
print(df['season'].value_counts().sort_index())

# --- 3D: MISSINGNESS AUDIT --------------------------------------------
print("\n=== 3D: MISSING DATA AUDIT ===")
print("Missing values per column:")
print(df[['Player', 'Team', 'season', 'G', 'MP', 'total_mp']].isna().sum())

os.makedirs('data/01-interim', exist_ok=True)
plt.figure(figsize=(12, 5))
sns.heatmap(df[['Player', 'Team', 'G', 'MP', 'total_mp', 'season']].isna(),
            cbar=False, yticklabels=False, cmap='viridis')
plt.title('Missing data -- NBA per-game stats (2017-18 through 2024-25)')
plt.tight_layout()
plt.savefig('data/01-interim/dataset3_missingness.png', dpi=150, bbox_inches='tight')
plt.show()

n_before = len(df)
df = df.dropna(subset=['Player', 'MP']).reset_index(drop=True)
print(f"Dropped {n_before - len(df)} rows missing Player or MP. Rows now: {len(df)}")

# --- 3E: OUTLIERS + SUSPICIOUS ENTRIES --------------------------------
print("\n=== 3E: OUTLIERS + SUSPICIOUS ENTRIES ===")

outlier_mp = df[df['MP'] > 48]
print(f"Rows with MP > 48 (impossible in regulation): {len(outlier_mp)}")
if len(outlier_mp) > 0:
    print(outlier_mp[['Player', 'Team', 'season', 'G', 'MP']].to_string(index=False))

low_games = df[df['G'] < 20]
print(f"\nPlayer-team stints with fewer than 20 games: {len(low_games)}")
print("  (excluded from top-3 ranking -- not genuine rotation players for that team)")

print("\nMedian MP per game by season:")
print(df.groupby('season')['MP'].median().round(1))

# --- 3F: TOP-3 MINUTES LEADERS LOOKUP TABLE --------------------------
print("\n=== 3F: TOP-3 MINUTES LEADERS LOOKUP TABLE ===")

# Require >= 20 games for that specific team stint to qualify
df_qualified = df[df['G'] >= 20].copy()
print(f"Player-team stints with >= 20 games (qualified): {len(df_qualified):,}")

df_qualified['rank_mp'] = (
    df_qualified
    .groupby(['season', 'Team'])['total_mp']
    .rank(method='first', ascending=False)
)

top3 = (
    df_qualified[df_qualified['rank_mp'] <= 3]
    [['season', 'Team', 'Player', 'G', 'MP', 'total_mp', 'rank_mp']]
    .sort_values(['season', 'Team', 'rank_mp'])
    .reset_index(drop=True)
)

expected = 30 * 8 * 3
print(f"\nTop-3 table shape: {top3.shape}  (expected {expected} rows)")
print(f"All 30 teams present every season: {(top3.groupby('season')['Team'].nunique() == 30).all()}")

print("\nSample -- 2024 Golden State Warriors:")
print(top3[(top3['Team'] == 'GSW') & (top3['season'] == 2024)].to_string(index=False))

print("\nSample -- 2020 LA Lakers:")
print(top3[(top3['Team'] == 'LAL') & (top3['season'] == 2020)].to_string(index=False))

# --- 3G: SAVE ---------------------------------------------------------
print("\n=== 3G: SAVE ===")
os.makedirs('data/02-processed', exist_ok=True)

full_path = 'data/02-processed/nba_player_stats_clean.csv'
df.to_csv(full_path, index=False)
print(f"Full cleaned stats table  ->  {full_path}  {df.shape}")

top3_path = 'data/02-processed/nba_top3_leaders_by_team_season.csv'
top3.to_csv(top3_path, index=False)
print(f"Top-3 lookup table        ->  {top3_path}  {top3.shape}")

print("\n=== CLEANING JUSTIFICATION ===")
print(
    "1. Dropped repeated BBRef header rows (Rk = 'Rk') -- formatting artifacts\n"
    "   from the HTML export, not real player rows.\n"
    "2. Dropped aggregate rows (TOT, 2TM, 3TM, 4TM) and kept per-team rows for\n"
    "   traded players. Each player's minutes are attributed to the specific team\n"
    "   they played for, so a traded player can qualify for the top-3 of each team\n"
    "   they contributed to based on their games played for that team specifically.\n"
    "3. Dropped rows missing Player name or MP -- no useful data for ranking.\n"
    "4. Top-3 ranking requires >= 20 games for that specific team stint -- excludes\n"
    "   short-term signings and 10-day contracts who are not genuine rotation players.\n"
    "5. Ranked by total_mp (MP x G), not per-game MP alone -- correctly weights\n"
    "   players who played many games over those with few high-minute appearances."
)

print(f"\nFinal full table : {df.shape}")
print(f"Final top-3 table: {top3.shape}")
```

NBA teams. The purpose of this dataset in our project is specifically to identify, 
for every team-season, the three players who led that team in total minutes played.

variables we use are:

- **Player:** Player name, used as the join key to Dataset #2 after light name 
standardization for special characters and suffixes.

- **Team:** The team abbreviation for that specific row (e.g. LAL, BOS, GSW). 
For traded players we keep one row per team they played for, so their minutes 
are attributed to each specific team rather than combined into a season total.

- **G (Games Played):** Number of regular-season games the player appeared in 
for that specific team. A player traded mid-season will have a separate G value 
for each team they played for. We require a minimum of 20 games for a player to 
qualify for a team's top-3 ranking, since anything below that is a short-term 
stint rather than a genuine rotation role.

- **MP (Minutes Per Game):** Average minutes played per game for that team. 
This is a per-game average, not a season total. To get total minutes we multiply 
MP by G.

- **total_mp (derived):** Total minutes played for that team, computed as MP 
times G. This is the variable we rank players by within each team-season to 
identify the top-3 minutes leaders. We use total minutes rather than per-game 
average because a player who averaged 38 minutes in 10 games should not outrank 
a player who played all 70 games at 30 minutes per game.

- **Awards:** A concatenated string of any honors the player received that season, 
for example AS for All-Star and NBA1, NBA2, NBA3 for All-NBA First, Second, and 
Third Team. We parse this column to flag players who are eligible under the 2023 
NBA Player Participation Policy, which defines a star as any player who received 
one of these honors in any of the three prior seasons.

The final top-3 lookup table contains exactly 720 rows, 3 players for each of 
the 30 teams across each of the 8 seasons, with all 30 teams represented in 
every season.

## Question 2b:
This dataset has two real concerns and one structural decision worth documenting.

The first concern is how we handle players traded mid-season. When a player is 
traded, Basketball Reference creates multiple rows: one per team they played for, 
plus aggregate rows labeled TOT, 2TM, or 3TM that combine their full-season 
stats. We drop all aggregate rows and keep only the per-team rows. This means 
each player's minutes are attributed to the specific team they played for, and a 
traded player can qualify for the top-3 of each team they played on, provided 
they meet the 20-game minimum for that specific team stint. Across our 8 seasons 
there are 610 such player-seasons split across 2 or more teams, which is why the 
raw data has 5,689 rows but only 5,079 after removing aggregate rows. The 
trade-off of this approach is that a player traded mid-season with 18 games on 
each team would be excluded from both teams' top-3 rankings since neither stint 
meets the 20-game threshold, even though they were a meaningful player for parts 
of both seasons. We accept this as a reasonable limitation since a player with 
fewer than 20 games on a team is genuinely not one of that team's primary 
rotation players for that season.

The second concern is the 20-game minimum threshold itself. We chose 20 games 
because it filters out short-term signings, 10-day contracts, and brief injury 
returns who are not genuine high-usage starters. The output shows 1,479 
player-team stints below this threshold, meaning roughly 29% of all per-team 
rows are excluded from the ranking. This is expected and correct behavior since 
many NBA roster spots are filled by players who appear in only a handful of games.

Finally, one non-issue worth naming: the dataset shows zero missing values across 
all key columns (Player, Team, G, MP, total_mp, season) after cleaning, confirmed 
by both the missingness table and the completely uniform heatmap. The dataset is 
spotless for the variables we care about.

### Block 5: Dimensionality Reduction via PCA

PCA on seven standardized team statistics (SRS, offensive/defensive ratings, pace, roster age, points scored/allowed per game) yields two principal components explaining 77.4% of variation. PC1 (40.5%) represents overall team quality, with offensive rating, defensive rating, and SRS loading heavily. PC2 (36.9%) captures secondary patterns, likely pace-related.

PC1 correlates extremely strongly with win percentage (r = 0.94) and playoff seeding (r = -0.91), while PC2 shows weaker relationships with outcomes.

```python
# Plot 6: PCA combines multiple stats into two main dimensions.
pcaColumns = [
    "srs", "offensive_rating", "defensive_rating",
    "pace", "roster_age", "points_scored_per_game",
    "points_allowed_per_game"
]

pcaData = df[pcaColumns].dropna()

scaler = StandardScaler()
scaledData = scaler.fit_transform(pcaData)

pcaModel = PCA(n_components=2)
pcaScores = pcaModel.fit_transform(scaledData)

pcaTable = df.loc[pcaData.index, ["team", "season", "win_pct", "playoff_seed"]].copy()
pcaTable["PC1"] = pcaScores[:, 0]
pcaTable["PC2"] = pcaScores[:, 1]

loadingTable = pd.DataFrame(
    pcaModel.components_.T,
    columns=["PC1", "PC2"],
    index=pcaColumns
)

plt.figure(figsize=(9, 7))
sns.scatterplot(data=pcaTable, x="PC1", y="PC2", hue="win_pct", palette="viridis", s=70)

for statName in loadingTable.index:
    plt.arrow(0, 0, loadingTable.loc[statName, "PC1"] * 3, loadingTable.loc[statName, "PC2"] * 3,
              color="black", alpha=0.7, head_width=0.06)
    plt.text(loadingTable.loc[statName, "PC1"] * 3.2, loadingTable.loc[statName, "PC2"] * 3.2,
             statName, fontsize=9)

plt.title("PCA Biplot of NBA Team Statistics")
plt.xlabel(f"PC1 ({pcaModel.explained_variance_ratio_[0] * 100:.1f}% variance)")
plt.ylabel(f"PC2 ({pcaModel.explained_variance_ratio_[1] * 100:.1f}% variance)")
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)
plt.tight_layout()
plt.show()

pc1Variance = pcaModel.explained_variance_ratio_[0] * 100
pc2Variance = pcaModel.explained_variance_ratio_[1] * 100

print(f"Analysis: PCA reduces the team stats into two main patterns, with PC1 explaining {pc1Variance:.1f}% of the variation and PC2 explaining {pc2Variance:.1f}%.")
```

### Block 6: Principal Component Analysis (PCA)

PCA reduces multiple correlated team statistics into a smaller number of principal components, helping us identify the main patterns that distinguish successful from unsuccessful teams. By combining variables like offensive rating, defensive rating, pace, and roster age, PCA reveals which dimensions matter most for understanding team performance. PC1 and PC2 together explain over 77% of the variation in team statistics, suggesting that team success can be largely understood through two main patterns.

```python
# Plot 7: PC1 compared to actual team outcomes.
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.regplot(data=pcaTable, x="PC1", y="win_pct", ax=axes[0], scatter_kws={"alpha": 0.6})
pc1WinR, pc1WinP = stats.pearsonr(pcaTable["PC1"], pcaTable["win_pct"])
axes[0].set_title(f"PC1 vs Win Percentage\nr = {pc1WinR:.2f}")
axes[0].set_xlabel("PC1")
axes[0].set_ylabel("Win Percentage")

sns.regplot(data=pcaTable, x="PC1", y="playoff_seed", ax=axes[1], scatter_kws={"alpha": 0.6})
pc1SeedR, pc1SeedP = stats.pearsonr(pcaTable["PC1"], pcaTable["playoff_seed"])
axes[1].set_title(f"PC1 vs Playoff Seed\nr = {pc1SeedR:.2f}")
axes[1].set_xlabel("PC1")
axes[1].set_ylabel("Playoff Seed")
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()

print(f"Analysis: PC1 strongly connects to team success because it relates strongly to win percentage (r = {pc1WinR:.2f}) and playoff seed (r = {pc1SeedR:.2f}).")
```

### Analysis Part 1: What Drives Team Success?

Our EDA establishes that team success in the NBA is dominated by a single underlying quality dimension, captured by PC1 (which combines SRS, offensive/defensive efficiency). This dimension explains 94% of variation in wins and 91% of variation in playoff seeding. 

This finding has critical implications for our research question: **Any association between strategic rest and team success may partly reflect that strong teams (high PC1) both rest more and win more due to superior talent.** Our control variable strategy using SRS, offensive rating, and defensive rating as controls is therefore essential. Without controlling for team strength, we cannot isolate whether rest independently predicts success or merely correlates with it.

Roster age and pace, while interesting, show weak direct relationships with wins. This validates our choice to include them as controls without expecting them to be primary drivers.

### Analysis Part 2: Pre/Post Policy Baseline Changes

Between the pre-policy era (2017-18 to 2022-23) and post-policy era (2023-24 to 2024-25), offensive rating increased significantly (p = 0.000) and defensive rating increased significantly (p = 0.000). However, pace showed no significant change (p = 0.248), and roster age showed no significant change (p = 0.492).

These findings suggest:
1. **The 2023 policy did not fundamentally alter how teams construct rosters or play at different speeds.** If the policy drove behavioral changes in rest patterns, those changes occurred without requiring teams to shift roster composition or game pace.
2. **Offensive and defensive rating increases reflect broader league evolution** (three-point emphasis, rule changes, evolved tactics) rather than policy-driven changes in load management strategy.
3. **Any detected policy effect on rest must come from changes in rest decisions themselves, not from secondary changes in team construction.** This is methodologically cleaner for Dataset 2 analysis.

### Analysis Part 3: Team Quality as the Baseline Model

PC1 serves as a nearly perfect predictor of team success (r = 0.94 with win %), leaving limited residual variance for other factors to explain. When we eventually introduce rest patterns from Dataset 2, the critical question becomes: **Do rest decisions add explanatory power beyond team strength?**

Put differently:
- **If rest adds no value:** Teams that rest more simply happen to be stronger teams (confounding).
- **If rest adds value:** Better teams achieve better results partly because they strategically rest while maintaining competitiveness.
- **If rest subtracts value:** Strong teams could achieve even better results by resting less (load management is counterproductive).

Our baseline model team strength predicts wins is extremely strong. Rest patterns must overcome this strong baseline to show meaningful independent effects.

### Section 2: Note Specificity Over Time

The `non_specific_note` flag identifies events where the notes field contains no meaningful reason (e.g., "placed on IL" with no further detail).
```python
relinquished = df2[df2['event_type'] == 'relinquished'].copy()

specificity = relinquished.groupby('season').agg(
    total=('non_specific_note', 'count'),
    non_specific=('non_specific_note', 'sum')
).reset_index()
specificity['pct_non_specific'] = specificity['non_specific'] / specificity['total'] * 100

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: line plot of non-specific % over time
axes[0].plot(specificity['season'], specificity['pct_non_specific'], marker='o', color='#e07b39', linewidth=2)
axes[0].axvline(2024, linestyle='--', color='gray', alpha=0.7, label='2023 Policy')
axes[0].set_title('% Non-Specific Notes by Season\n(Relinquish Events Only)', fontsize=12)
axes[0].set_xlabel('Season (End Year)', fontsize=11)
axes[0].set_ylabel('% Non-Specific Notes', fontsize=11)
axes[0].legend()

# Right: top 15 most common notes
top_notes = relinquished['notes'].value_counts().head(15)
axes[1].barh(top_notes.index[::-1], top_notes.values[::-1], color='#5b9bd5')
axes[1].set_title('Top 15 Most Common Notes\n(Relinquish Events)', fontsize=12)
axes[1].set_xlabel('Count', fontsize=11)
axes[1].set_ylabel('Note Text', fontsize=11)

plt.tight_layout()
plt.show()

print(specificity[['season','total','non_specific','pct_non_specific']].to_string(index=False))
```

### Section 3: Narrow vs. Broad Rest Detection

The NBA's 2023 Player Participation Policy fines teams for resting star players in nationally televised games. This creates an incentive to reclassify strategic rest as minor injury. We construct two rest definitions to detect this:
- **Narrow**: explicit keywords — *rest, load management, not injury related, precautionary*
- **Broad**: narrow + vague single-session notes — *soreness, tightness, fatigue, maintenance*

A widening gap between the two post-2023 is consistent with teams substituting explicit rest labels for vague injury language.

```python
narrow_keywords = ['rest', 'load management', 'not injury related', 'precautionary']
broad_keywords = narrow_keywords + ['soreness', 'sore', 'tightness', 'tight', 'fatigue', 'maintenance', 'stiffness']

def flag_rest(notes, keywords):
    if pd.isna(notes):
        return False
    notes_lower = notes.lower()
    return any(kw in notes_lower for kw in keywords)

relinquished = relinquished.copy()
relinquished['narrow_rest'] = relinquished['notes'].apply(lambda x: flag_rest(x, narrow_keywords))
relinquished['broad_rest'] = relinquished['notes'].apply(lambda x: flag_rest(x, broad_keywords))

rest_by_season = relinquished.groupby('season').agg(
    total=('narrow_rest', 'count'),
    narrow=('narrow_rest', 'sum'),
    broad=('broad_rest', 'sum')
).reset_index()
rest_by_season['narrow_rate'] = rest_by_season['narrow'] / rest_by_season['total'] * 100
rest_by_season['broad_rate'] = rest_by_season['broad'] / rest_by_season['total'] * 100

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(rest_by_season['season'], rest_by_season['narrow_rate'], marker='o', label='Narrow (explicit rest)', color='#2c7bb6', linewidth=2)
ax.plot(rest_by_season['season'], rest_by_season['broad_rate'], marker='s', label='Broad (+ vague injury terms)', color='#d7191c', linewidth=2, linestyle='--')
ax.axvline(2024, linestyle=':', color='gray', alpha=0.8, label='2023 Policy effective')
ax.set_title('Rest Detection Rate by Season: Narrow vs. Broad Definition', fontsize=13)
ax.set_xlabel('Season (End Year)', fontsize=11)
ax.set_ylabel('% of Relinquish Events Flagged as Rest', fontsize=11)
ax.legend()
plt.tight_layout()
plt.show()

print(rest_by_season[['season','narrow_rate','broad_rate']].to_string(index=False))
```

The narrow rest rate is near zero across almost all seasons, indicating that teams almost never 
use explicit rest language in transaction notes — even before the 2023 policy. The only exception 
is 2017-18 (5.3%), which likely reflects an early period before teams became aware of the 
reputational risk of explicit rest labeling. The broad rate by contrast stays consistently 
elevated at 13–20%, reflecting the steady use of vague injury language like soreness and 
tightness throughout the window. This near-zero narrow rate is itself a key finding: it means 
our rest proxy must rely on the broad definition, and that the 2023 policy did not meaningfully 
change an already-established pattern of avoiding explicit rest labels.

### Section 4: Absence Duration Distribution

The dataset records when players are placed on and activated from the inactive list, but does not directly state how many games were missed. We approximate absence duration by computing the gap in days between paired relinquish and acquire events for the same player and team within a season.

```python
relinquish_events = df2[df2['event_type'] == 'relinquished'][['date','team','player','season']].copy()
acquire_events = df2[df2['event_type'] == 'acquired'][['date','team','player','season']].copy()

relinquish_events = relinquish_events.sort_values('date')
acquire_events = acquire_events.sort_values('date')

paired = pd.merge_asof(
    relinquish_events,
    acquire_events.rename(columns={'date': 'return_date'}),
    left_on='date', right_on='return_date',
    by=['team', 'player', 'season'],
    direction='forward'
)

paired['gap_days'] = (paired['return_date'] - paired['date']).dt.days
paired_valid = paired.dropna(subset=['gap_days'])
paired_valid = paired_valid[paired_valid['gap_days'] > 0]

fig, ax = plt.subplots(figsize=(10, 5))
plot_data = paired_valid[paired_valid['gap_days'] <= 60]['gap_days']
ax.hist(plot_data, bins=30, color='#5b9bd5', edgecolor='white')
ax.set_title('Distribution of Absence Duration (Gap Days)\nTruncated at 60 Days for Readability', fontsize=13)
ax.set_xlabel('Days Between Relinquish and Acquire Events', fontsize=11)
ax.set_ylabel('Number of Events', fontsize=11)
ax.axvline(paired_valid['gap_days'].median(), color='#d9534f', linestyle='--', label=f"Median: {paired_valid['gap_days'].median():.0f} days")
ax.legend()
plt.tight_layout()
plt.show()

q1 = paired_valid['gap_days'].quantile(0.25)
q3 = paired_valid['gap_days'].quantile(0.75)
print(f"Median gap: {paired_valid['gap_days'].median():.0f} days")
print(f"IQR: {q1:.0f}–{q3:.0f} days")
print(f"Events > 30 days (likely true injuries): {(paired_valid['gap_days'] > 30).sum()}")
print(f"Events <= 3 days (likely rest): {(paired_valid['gap_days'] <= 3).sum()}")
```

The distribution is strongly right-skewed, so median and IQR are the appropriate summary statistics here. Short absences (1–3 days) form the dominant cluster and are the most likely candidates for strategic rest. The long right tail represents season-ending or multi-week injuries. Full games-missed calculation requires joining with the game schedule from Dataset #1, which will be completed at the analysis stage.

### Section 5: Team-Level Rest Rate Heatmap

Having characterized the dataset's limitations, we now examine team-level variation in broad rest rates across seasons. This provides a preview of the primary explanatory variable in our regression analysis.

```python
team_rest = relinquished.copy()
team_rest['broad_rest'] = team_rest['notes'].apply(lambda x: flag_rest(x, broad_keywords))

team_season = team_rest.groupby(['team','season']).agg(
    total=('broad_rest','count'),
    rest_events=('broad_rest','sum')
).reset_index()
team_season['rest_rate'] = team_season['rest_events'] / team_season['total'] * 100

pivot = team_season.pivot(index='team', columns='season', values='rest_rate')

fig, ax = plt.subplots(figsize=(14, 12))
sns.heatmap(pivot, cmap='YlOrRd', linewidths=0.5, annot=True, fmt='.0f',
            cbar_kws={'label': 'Rest Rate (% of relinquish events)'}, ax=ax)
ax.set_title('Broad Rest Rate by Team and Season (%)\nHigher = More Events Flagged as Strategic Rest', fontsize=13)
ax.set_xlabel('Season (End Year)', fontsize=11)
ax.set_ylabel('Team', fontsize=11)
plt.tight_layout()
plt.show()
```

The Grizzlies show the most persistently elevated broad rest rate across all 8 seasons, 
ranging from 13% to 39%, suggesting a long-standing organizational philosophy around 
player management. The Spurs similarly remain elevated throughout (10–33%), consistent 
with their historically conservative approach to veteran workloads. The Pistons spike to 
42% in 2022–23 — the highest single value in the dataset — which likely reflects roster 
management during a rebuilding year rather than traditional load management of star players.

Notably, several teams show increases in the 2022–23 season specifically (Bucks 33%, 
Raptors 30%, Pacers 32%), the season just before the 2023 policy took effect, which may 
reflect a final surge in rest usage before fines were introduced. Post-2024, many teams 
show declining or zero values — though as established in Section 3, this likely reflects 
label substitution rather than a genuine reduction in rest behavior.

This team-level variation in broad rest rate is the primary explanatory variable that will 
be joined with Dataset #1 win percentages and playoff seedings in the regression analysis.

## Exploratory Data Analysis for Dataset #3: NBA Player Per-Game Statistics (2017–2025)

This section explores individual player performance data sourced from Basketball Reference covering the 2017-18 through 2024-25 seasons. The dataset contains 5,079 player-team-season records with columns for player name, team, games played, minutes per game, and award designations (All-Star, All-NBA). We derive `total_mp` (minutes per game × games played) to rank and identify each team's top-3 minutes leaders across the 8 seasons. Our goal is to validate that the top-3 selection captures genuine high-usage players and characterize their usage patterns over time as a foundation for analyzing strategic rest practices.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset 3 cleaned data
df_full = pd.read_csv('data/02-processed/nba_player_stats_clean.csv')
df_top3 = pd.read_csv('data/02-processed/nba_top3_leaders_by_team_season.csv')

print(f"Full stats: {df_full.shape}")
print(f"Top-3 lookup: {df_top3.shape}")
print(df_top3.head())
```

```python
# Minutes per game distribution across all players, all seasons
plt.figure(figsize=(10, 5))
plt.hist(df_full['MP'], bins=30, color='steelblue', edgecolor='black', alpha=0.7)
plt.xlabel('Minutes Per Game')
plt.ylabel('Frequency')
plt.title('Distribution of Minutes Per Game (All Players, 2017-18 to 2024-25)')
plt.axvline(df_full['MP'].median(), color='red', linestyle='--', linewidth=2, label=f'Median: {df_full["MP"].median():.1f}')
plt.legend()
plt.tight_layout()
plt.show()
```

### Plot 1: Minutes Per Game Distribution

Minutes per game across all players ranges from near 0 to the maximum (~36-40), with a median around 18-19 minutes. The distribution is right-skewed, reflecting that most players are reserves who see limited minutes while a smaller group of stars dominate playing time. This confirms the dataset captures the full roster and that top-3 selection based on total_mp (MP × Games) will correctly identify high-usage players.

```python
# Compare total minutes for top-3 leaders vs all other players
top3_players = set(df_top3['Player'])
df_full['is_top3'] = df_full['Player'].isin(top3_players)

plt.figure(figsize=(10, 6))
sns.violinplot(data=df_full, x='is_top3', y='total_mp', palette=['lightblue', 'darkblue'])
plt.xlabel('Top-3 Minutes Leader')
plt.ylabel('Total Minutes (MP × G)')
plt.title('Total Minutes: Top-3 Leaders vs Rest of Roster')
plt.xticks([0, 1], ['Rest of Roster', 'Top-3 Leaders'])
plt.tight_layout()
plt.show()

print(f"Top-3 median total_mp: {df_full[df_full['is_top3']]['total_mp'].median():.0f}")
print(f"Rest median total_mp: {df_full[~df_full['is_top3']]['total_mp'].median():.0f}")
```

### Plot 2: Top-3 Leaders vs Rest of Roster

Top-3 minutes leaders (median ~2,300 total minutes) have substantially higher usage than the rest of the roster (median ~200 total minutes). This clear separation validates that the top-3 selection correctly identifies the team's primary players and captures genuine high-volume rotations, not borderline cases.

```python
# Average total_mp for top-3 leaders per season
trend_data = df_top3.groupby('season')['total_mp'].mean()

plt.figure(figsize=(10, 5))
plt.plot(trend_data.index, trend_data.values, marker='o', linewidth=2, markersize=8, color='steelblue')
plt.xlabel('Season')
plt.ylabel('Average Total Minutes')
plt.title('Average Total Minutes: Top-3 Leaders by Season')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print(trend_data)
```

### Plot 3: Top-3 Total Minutes Over Time

Average total minutes for top-3 leaders remains relatively stable across the 2017-18 through 2024-25 seasons (~2,200-2,300 total minutes), with no dramatic increases or decreases. This stability suggests that teams have maintained consistent reliance on their star players across the period, providing a stable baseline for analyzing rest patterns.
