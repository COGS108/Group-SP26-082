## Research Question


Research Question:

Among NBA teams from the 2016–17 through 2024–25 regular seasons, is the rate of strategic rest games among a team's top-3 minutes leaders associated with regular season win percentage and playoff seeding, after controlling for team strength (preseason over/under win total), roster age, and back-to-back game frequency? ## Background and Prior Work



The practice of "load management" — strategically resting healthy NBA players during the regular season to preserve them for the playoffs — has become one of the most contentious topics in modern professional basketball. Once an obscure concept used by a small number of teams, load management entered mainstream attention when Kawhi Leonard sat out 22 of 82 regular season games during the Toronto Raptors' 2018-19 championship season, an approach widely credited with helping him win Finals MVP.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Since then, the practice has spread league-wide, generating concerns from broadcasters, fans, and the league office about competitive integrity and the value of regular-season games. The fundamental question for our project is whether this strategy is associated with team success, and whether the league's recent attempt to curb the practice has changed team behavior.

The empirical literature on regular-season rest and player outcomes is small but growing, and notably its findings have not strongly supported the assumed benefits of load management. Belk and colleagues conducted one of the earliest systematic analyses, examining 811 NBA players over the 2005-06 through 2014-15 seasons using DNP-CD ("did not play – coach's decision") records from Basketball Reference and Pro Sports Transactions.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Comparing players who missed fewer than 5 regular-season games due to rest against those who missed 5 to 9 games, they found no significant differences in playoff performance metrics or injury risk. The authors concluded that, contrary to widespread belief, regular-season rest did not improve playoff outcomes for the rested players themselves.

A more recent study by Pradhan and Miller extended this work using data from the 2016-17 through 2020-21 seasons, with rest decisions sourced from Spotrac.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Using Player Efficiency Rating and Win Shares as outcome measures, they similarly found insufficient evidence that strategic rest during the regular season translated into better post-season performance. A nested case-control comparison of 184 players (92 cases of rested players matched with 92 controls) showed no meaningful differences in performance changes from regular to post-season. Importantly, both studies focused on individual player performance rather than team-level outcomes such as win percentage or playoff seeding, and neither examined how the practice's prevalence has shifted over time. Our project addresses these gaps directly.

In response to growing concerns about load management's effect on regular-season competitiveness and broadcast value, the NBA Board of Governors in September 2023 approved the Player Participation Policy, which took effect at the start of the 2023-24 season.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) The policy formally defines a "star player" as any player named to an All-Star or All-NBA team in any of the prior three seasons, and prohibits teams from resting these players in nationally televised games or in-season tournament games without approved justification. Violations carry escalating fines beginning at USD 100,000 for a first offense and rising to USD 1.25 million for a third. Bobby Marks of ESPN provided detailed coverage of the policy's mechanics, including the specific rest restrictions, exception categories for older or high-mileage stars, and the league's investigative authority over suspect injury designations.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) This policy creates a natural setting for examining whether legal restrictions on rest decisions actually change team behavior — a question that has not yet been studied empirically given that the policy is recent.

Our project builds on the prior literature in three specific ways. First, we shift the unit of analysis from individual players to teams, asking whether a team's overall rest patterns are associated with regular-season win percentage and playoff seeding rather than individual player performance. Second, we extend the time window through the 2024-25 season, capturing both the pre- and post-policy eras. Third, by leveraging the policy's binary nature (it applies only to players meeting specific All-Star or All-NBA criteria), we can compare rest patterns for policy-eligible stars against those of non-star high-minutes players over the same period. This comparison group acts as a baseline to help separate the policy's effect from any general league-wide trends in rest behavior.

2a. 
- **W/L%(Win Percentage):** Simply outputs how many games a team won out of 82 regular season games. A .700 means they 
won 70% of their games (typically a really good team). While a .500 is perfectly average, below that 
means more losses than wins.

- **PS/G (Points Scored Per Game):** How many points the team averages per game. For context, in today's 
NBA anything above 115 is considered a high-scoring offense.


- **PA/G (Points Allowed/ Conceded Per Game):** This statistic represents the total number of points a team conceded in a single season. Lower the points conceded the better. (If Points scored > Points Conceded = Win), a good defensive teams usually sit around 105-110.
- **SRS (Simple Rating System):** is a team's overall rating that accounts for both how much they win by and who they played against. 0 means perfectly average, that team is right in the middle of the league. A +8 means that team is significantly better than average. Think of it like a point spread, a team with an SRS of +8 would be expected to beat an average team by 8 points on a neutral court.
-8 means one of the worst.

  
- **ORtg (Offensive Rating):** accounts for the points scored per 100 possessions. It's a more accurate statistic that accounts for the pacing of the game. League average is around 108-112.
- **DRtg (Defensive Rating)** accounts for the points allowed per 100 possessions. Lower the better, a true reflection of an efficient defense.

- **Pace:** How fast a team plays, basically possessions per game. High pace teams play up tempo, while low pace teams slow it down.

  
- **Age:** Average age of the roster. Most NBA rosters average around 25-28 years old. This matters for our project because older rosters tend to need more rest.

## Question 2a:
Dataset two is a player-level event log compiled from ProSportsTransactions and distributed on Kaggle by user jacquesoberweis. Every row represents a single roster transaction, a player either being placed on the inactive list / missing a game, or being activated back into the lineup. After tidying and restricting to our project window (2017-18 through 2024-25), we are left with roughly 15,000 events covering 994 unique players across all 30 NBA teams. The key variables we will use are:

- **Date:** The calendar date the transaction was reported. We parse this from the raw string and use it to assign each event to an NBA season.

- **Team:** The team mascot (e.g. Lakers, Celtics) associated with the transaction. We use this as the merge key with Dataset #1 after standardizing to full team names.

- **Acquired:** The name of a player being activated back into the lineup. Populated only on return events, otherwise empty by design. Present in roughly 47% of rows.

- **Relinquished:** The name of a player being placed on the inactive list or missing a game. Populated only on absence events, otherwise empty by design. Present in roughly 53% of rows. The mutual exclusivity of acquired and relinquished is not a missingness problem, every row contains exactly one of the two.

- **Notes:** A short free-text description of the reason for the transaction. This is the most important column for our analysis. Values range from highly specific (placed on IL with sprained left ankle) to non-specific (placed on IL, DNP). We will keyword-filter this field for terms such as rest, soreness, sprain, tightness, sore, fatigue, and precautionary to build our strategic-rest proxy. The 522 events labeled with NBA health and safety protocols are COVID-19 absences concentrated in the 2020-21 season and are flagged separately so they do not contaminate the rest proxy.


Derived variables we have constructed from the raw fields:

- **Event_type:** Labels whether the row is a relinquish (player going out) or acquire (player coming back). Derived from which of the two name fields is populated.

- **Player:** A single unified player-identifier column, taken from relinquished on absence events and acquired on return events. This allows clean groupby operations regardless of event direction.

- **Season:** NBA season label following the same convention as Dataset #1 (the year in which the season ends, so games played Oct 2023 to Apr 2024 are labeled 2024). Computed by adding 1 to the year for October-December events and using the year directly for January-September events.

- **Non_specific_note:** A boolean flag for entries whose notes are uninformative (placed on IL, DNP, or empty). Useful for downstream sensitivity analysis.

## Question 2b:
This dataset has four substantial concerns that we have already begun to mitigate but that we will continue to flag throughout the analysis.

The first concern is that the 2024-25 season is incomplete. The Kaggle snapshot ends on 2025-01-12, so our final season contains only 823 events, roughly the first half. Any season-level rates we compute for 2024-25 must be either restricted to that same date range across all seasons for fair pre/post-policy comparison, or treated as provisional. We will note this prominently in the EDA checkpoint and likely supplement with a more recent ProSportsTransactions scrape if available.

The second concern is that the reporting convention for the reason field has changed over the project window. The share of completely non-specific note entries (e.g. bare placed on IL) drops monotonically from 44.0% in 2017-18 to essentially 0% by 2023-24. This is almost certainly a reporting change, not a real behavioral change, teams or the data compiler increasingly always include a stated reason. The practical consequence is that the early seasons of our window are less informative about why a player sat out, which mechanically suppresses our rest-detection rate in those seasons. We will resolve this in the EDA checkpoint by plotting rest-detection rates under both inclusion rules across seasons and selecting the primary analysis window based on where rest detection appears stable, while running the full window as a sensitivity check.

The third concern is that strategic rest is fundamentally unobservable in the labels. Even when a reason is given, teams have no obligation to label something as rest, and after the 2023 Player Participation Policy specifically, they have a financial incentive not to. We expect post-2023 events labeled as minor injury (soreness, sprain, tightness, single-game absences) to be partially mislabeled rest. This means our rest-rate variable systematically undercounts true strategic rest in the post-policy era, which is exactly the era our secondary research question is asking about. We will address this with a minor-injury proxy that includes short, single-game soreness-style absences as candidate rest events.

The fourth concern is that the dataset does not include games played or minutes played. It only records that a transaction occurred, not how many games a player missed as a result. To convert a relinquish event into games missed, we will join against the team game-log calendar from Dataset #1 and count regular-season games occurring between each player's relinquish and their next paired acquire event. Players never explicitly re-acquired in the data are right-censored and will be handled as such.

Finally, two minor non-issues worth naming explicitly so they are not mistaken for data quality problems. First, the roughly 7,000 empty values in each of acquired and relinquished are structural, not missing data, every row is exactly one event type. Second, two rows had no player name in either field and were dropped, leaving 15,008 usable events.

```python
# =====================================================================
# 3A–3D: Load, Tidy, Audit, and Clean — NBA Injury Event Log (2017–2025)
# =====================================================================
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- 3A: LOAD RAW DATA -----------------------------------------------
df = pd.read_csv('data/00-raw/nba_injury_events_2016-2025.csv')
print("=== 3A: RAW LOAD ===")
print(f"Raw shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print("\nFirst 5 rows:")
print(df.head())

# --- 3B: TIDY --------------------------------------------------------
# Standardize column names
df.columns = [c.strip().lower() for c in df.columns]

# Parse dates
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# Strip whitespace, normalize empty strings to NaN
for col in ['team', 'acquired', 'relinquished', 'notes']:
    df[col] = df[col].astype(str).str.strip().replace({'nan': np.nan, '': np.nan})

# Derive event_type (relinquish = going out, acquire = coming back)
df['event_type'] = np.where(df['relinquished'].notna(), 'relinquished', 'acquired')
df['player']     = df['relinquished'].fillna(df['acquired'])

# Assign NBA season label (Oct–Apr; labeled by ending year to match Dataset #1)
def assign_season(d):
    if pd.isna(d): return np.nan
    return d.year + 1 if d.month >= 10 else d.year
df['season'] = df['date'].apply(assign_season).astype('Int64')

# Restrict to project window: 2017-18 through 2024-25
df = df[df['season'].between(2018, 2025)].reset_index(drop=True)

print("\n=== 3B: TIDIED ===")
print("Each row = one roster transaction. Each column = one variable. Data is tidy.")
print(f"Tidied shape: {df.shape}")
print(df.head())

# --- 3C: SIZE --------------------------------------------------------
print("\n=== 3C: DATASET SIZE ===")
print(f"Total events:    {len(df):,}")
print(f"Unique players:  {df['player'].nunique():,}")
print(f"Unique teams:    {df['team'].nunique()}")
print(f"Date range:      {df['date'].min().date()} → {df['date'].max().date()}")
print("\nEvents per season:")
print(df['season'].value_counts().sort_index())
print("\nEvents by type:")
print(df['event_type'].value_counts())

# --- 3D: MISSINGNESS AUDIT + CLEAN ----------------------------------
print("\n=== 3D: MISSING DATA AUDIT ===")
print("Missing values per column:")
print(df.isna().sum())
print("\nNote: 'acquired' and 'relinquished' are mutually exclusive by design —")
print("each row holds exactly one of them, so the apparent missingness is structural,")
print("not a data quality problem.")

# Visualize missingness pattern
os.makedirs('data/01-interim', exist_ok=True)
plt.figure(figsize=(10, 5))
sns.heatmap(df.isna(), cbar=False, yticklabels=False, cmap='viridis')
plt.title('Missing data pattern — NBA injury event log (2017–18 through 2024–25)')
plt.tight_layout()
plt.savefig('data/01-interim/dataset2_missingness.png', dpi=150, bbox_inches='tight')
plt.show()

# --- Cleaning Step 1: drop rows with no player attached (genuinely unusable) ---
n_before = len(df)
bad_rows = df[df['player'].isna()]
print(f"\nDropping {len(bad_rows)} rows with no player attached (unusable for analysis):")
print(bad_rows)
df = df.dropna(subset=['player']).reset_index(drop=True)
print(f"Rows: {n_before} → {len(df)}")

# --- Cleaning Step 2: normalize slashed multi-name entries ---
# e.g. "Maurice Ndour / Maurice N'dour" → "Maurice Ndour"
df['player'] = df['player'].astype(str).str.split(' / ').str[0]

# --- Cleaning Step 3: flag non-specific notes (proxy signal for strategic rest) ---
def is_non_specific(note):
    if pd.isna(note): return True
    return str(note).strip().lower() in ['placed on il', 'placed on il.', 'dnp']
df['non_specific_note'] = df['notes'].apply(is_non_specific)

# --- Reporting trend: share of non-specific notes by season ---
rel = df[df['event_type'] == 'relinquished']
print("\nShare of NON-SPECIFIC notes by season (relinquish events only):")
print(rel.groupby('season')['non_specific_note'].mean().round(3))
print("\nTop 10 most common note values:")
print(rel['notes'].value_counts().head(10))

# --- Cleaning justification (printed for clarity) -------------------
print("\n=== CLEANING JUSTIFICATION ===")
print(
    "1. Dropped 2 rows with missing player (no player name in either field) — "
    "these cannot be attributed to any player and are unusable.\n"
    "2. KEPT rows with NaN in 'acquired' or 'relinquished' — this is structural, "
    "not missing data (each row is exactly one event type).\n"
    "3. Resolved slashed multi-name entries to the first variant for stable "
    "player IDs across the dataset.\n"
    "4. KEPT rows regardless of note specificity — non-specific notes are "
    "themselves an informative signal for the strategic-rest classifier."
)

# --- Save cleaned dataset -------------------------------------------
os.makedirs('data/02-processed', exist_ok=True)
out_path = 'data/02-processed/nba_injury_events_clean.csv'
df.to_csv(out_path, index=False)
print(f"\n✓ Final cleaned shape: {df.shape}")
print(f"✓ Saved to: {out_path}")

# 3E. Flag outliers and suspicious entries
print("\n3E. Duplicate event check")
dup_columns = ['player', 'team', 'date', 'event_type']

duplicate_groups = df.duplicated(subset=dup_columns, keep='first').sum()
duplicate_total = df.duplicated(subset=dup_columns, keep=False).sum()
print(f"Exact duplicate groups: {duplicate_groups}")
print(f"Total duplicate rows (including repeated copies): {duplicate_total}")

if duplicate_total > 0:
    print("\nSample duplicate rows:")
    print(df[df.duplicated(subset=dup_columns, keep=False)][dup_columns + ['notes']].head(10).to_string(index=False))

print("\n3E. Relinquish→Acquire gap analysis")
rel = df[df['event_type'] == 'relinquished'].copy()
acq = df[df['event_type'] == 'acquired'].copy()

rel = rel.rename(columns={'date': 'date_rel'})
acq = acq.rename(columns={'date': 'date_acq'})

rel = rel.sort_values(['player', 'season', 'team', 'date_rel']).reset_index(drop=True)
acq = acq.sort_values(['player', 'season', 'team', 'date_acq']).reset_index(drop=True)

paired = pd.merge(
    rel,
    acq,
    on=['player', 'season', 'team'],
    suffixes=('_rel', '_acq')
)
paired = paired[paired['date_acq'] >= paired['date_rel']].copy()
paired = paired.sort_values(['player', 'season', 'team', 'date_rel', 'date_acq'])
paired = paired.groupby(['player', 'season', 'team', 'date_rel'], as_index=False).first()

paired['gap_days'] = (paired['date_acq'] - paired['date_rel']).dt.days

long_gaps = paired[(paired['gap_days'].notna()) & (paired['gap_days'] > 100)].copy()
print(f"Relinquish→Acquire pairs with gap > 100 days: {len(long_gaps)}")
if len(long_gaps) > 0:
    print(long_gaps[['player', 'team', 'season', 'date_rel', 'date_acq', 'gap_days']].head(10).to_string(index=False))
else:
    print("No large gaps over 100 days were found.")

print("\nTop 10 most suspicious/unusual entries:")
if len(long_gaps) > 0:
    suspicious_entries = long_gaps[['player', 'team', 'season', 'date_rel', 'date_acq', 'gap_days']].copy()
else:
    suspicious_entries = paired.sort_values('gap_days', ascending=False)[['player', 'team', 'season', 'date_rel', 'date_acq', 'gap_days']].head(10).copy()
print(suspicious_entries.head(10).to_string(index=False))

# 3F. Any remaining cleaning
print("\n3F. Remaining cleaning")
if duplicate_groups > 0:
    before_dedup = len(df)
    df = df.drop_duplicates(subset=dup_columns, keep='first').reset_index(drop=True)
    after_dedup = len(df)
    print(f"Dropped {before_dedup - after_dedup} exact duplicate rows.")
else:
    print("No exact duplicate rows to drop.")

# 3G. Save and confirm
print("\n3G. Save final cleaned dataset")
output_path = 'data/02-processed/nba_injury_events_clean.csv'
df.to_csv(output_path, index=False)
print(f"Overwrote: {output_path}")
print(f"Final shape: {df.shape}")
print("\nFinal sample rows:")
print(df.head(10).to_string(index=False))
```
