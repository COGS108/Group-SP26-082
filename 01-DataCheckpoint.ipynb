## Research Question


Research Question:

Among NBA teams from the 2017–18 through 2024–25 regular seasons, is the rate of strategic rest games among a team's top-3 minutes leaders associated with regular season win percentage and playoff seeding, after controlling for team strength (preseason over/under win total), roster age, and back-to-back game frequency? ## Background and Prior Work



The practice of "load management" — strategically resting healthy NBA players during the regular season to preserve them for the playoffs — has become one of the most contentious topics in modern professional basketball. Once an obscure concept used by a small number of teams, load management entered mainstream attention when Kawhi Leonard sat out 22 of 82 regular season games during the Toronto Raptors' 2018-19 championship season, an approach widely credited with helping him win Finals MVP.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Since then, the practice has spread league-wide, generating concerns from broadcasters, fans, and the league office about competitive integrity and the value of regular-season games. The fundamental question for our project is whether this strategy is associated with team success, and whether the league's recent attempt to curb the practice has changed team behavior.

The empirical literature on regular-season rest and player outcomes is small but growing, and notably its findings have not strongly supported the assumed benefits of load management. Belk and colleagues conducted one of the earliest systematic analyses, examining 811 NBA players over the 2005-06 through 2014-15 seasons using DNP-CD ("did not play – coach's decision") records from Basketball Reference and Pro Sports Transactions.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Comparing players who missed fewer than 5 regular-season games due to rest against those who missed 5 to 9 games, they found no significant differences in playoff performance metrics or injury risk. The authors concluded that, contrary to widespread belief, regular-season rest did not improve playoff outcomes for the rested players themselves.

A more recent study by Pradhan and Miller extended this work using data from the 2016-17 through 2020-21 seasons, with rest decisions sourced from Spotrac.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Using Player Efficiency Rating and Win Shares as outcome measures, they similarly found insufficient evidence that strategic rest during the regular season translated into better post-season performance. A nested case-control comparison of 184 players (92 cases of rested players matched with 92 controls) showed no meaningful differences in performance changes from regular to post-season. Importantly, both studies focused on individual player performance rather than team-level outcomes such as win percentage or playoff seeding, and neither examined how the practice's prevalence has shifted over time. Our project addresses these gaps directly.

In response to growing concerns about load management's effect on regular-season competitiveness and broadcast value, the NBA Board of Governors in September 2023 approved the Player Participation Policy, which took effect at the start of the 2023-24 season.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) The policy formally defines a "star player" as any player named to an All-Star or All-NBA team in any of the prior three seasons, and prohibits teams from resting these players in nationally televised games or in-season tournament games without approved justification. Violations carry escalating fines beginning at USD 100,000 for a first offense and rising to USD 1.25 million for a third. Bobby Marks of ESPN provided detailed coverage of the policy's mechanics, including the specific rest restrictions, exception categories for older or high-mileage stars, and the league's investigative authority over suspect injury designations.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) This policy creates a natural setting for examining whether legal restrictions on rest decisions actually change team behavior — a question that has not yet been studied empirically given that the policy is recent.

Our project builds on the prior literature in three specific ways. First, we shift the unit of analysis from individual players to teams, asking whether a team's overall rest patterns are associated with regular-season win percentage and playoff seeding rather than individual player performance. Second, we extend the time window through the 2024-25 season, capturing both the pre- and post-policy eras. Third, by leveraging the policy's binary nature (it applies only to players meeting specific All-Star or All-NBA criteria), we can compare rest patterns for policy-eligible stars against those of non-star high-minutes players over the same period. This comparison group acts as a baseline to help separate the policy's effect from any general league-wide trends in rest behavior.

## Question 2a:
Dataset one (NBA regular season game logs) quantifies statistics for each NBA season from 2017–18 through 2024–25. The key variables we are using and what they mean are:

- **W/L%(Win Percentage):** Simply outputs how many games a team won out of 82 regular season games. A .700 means they 
won 70% of their games (typically a really good team). While a .500 is perfectly average, below that 
means more losses than wins.

- **PS/G (Points Scored Per Game):** How many points the team averages per game. For context, in today's 
NBA anything above 115 is considered a high-scoring offense.


- **PA/G (Points Allowed/ Conceded Per Game):** This statistic represents the total number of points a team conceded in a single season. Lower the points conceded the better. (If Points scored > Points Conceded = Win), a good defensive teams usually sit around 105-110.
- **SRS (Simple Rating System):** is a team's overall rating that accounts for both how much they win by and who they played against. 0 means perfectly average, that team is right in the middle of the league. A +8 means that team is significantly better than average. Think of it like a point spread, a team with an SRS of +8 would be expected to beat an average team by 8 points on a neutral court.
-8 means one of the worst.

  
- **ORtg (Offensive Rating):** accounts for the points scored per 100 possessions. It's a more accurate statistic that accounts for the pacing of the game. League average is around 108-112.
- **DRtg (Defensive Rating)** accounts for the points allowed per 100 possessions. Lower the better, a true reflection of an efficient defense.

- **Pace:** How fast a team plays, basically possessions per game. High pace teams play up tempo, while low pace teams slow it down.

  
- **Age:** Average age of the roster. Most NBA rosters average around 25-28 years old. This matters for our project because older rosters tend to need more rest.

### Dataset #2: NBA Team Season Statistics (2017-18 to 2024-25)

**Dataset Description:**

This dataset contains comprehensive team-level statistics for each NBA season from 2017-18 through 2024-25, including win-loss records, efficiency ratings, pace of play, roster age, and playoff performance. These aggregate season statistics are essential for analyzing whether strategic rest patterns correlate with team-level outcomes (regular season win percentage and playoff seeding). Unlike game-level data, this dataset provides the outcome variables and control variables needed for regression analysis.

**Key Variables:**
- **W/L%:** Win-loss percentage (primary outcome variable)
- **Playoff Seeding/Status:** Regular season finish position and playoff participation (secondary outcome)
- **ORtg & DRtg:** Offensive and defensive efficiency ratings (control variables for team strength)
- **Pace:** Possessions per game (proxy for back-to-back frequency)
- **Age:** Average roster age (control variable - older rosters need more rest)
- **PW/PL:** Playoff wins and losses (secondary outcome variable)

**Data Concerns:**
- Historical completeness: Some years may have incomplete or revised statistics
- Playoff data only available for teams that made postseason
- Efficiency ratings depend on correct game counting and can have edge cases
- Younger age of data collection in early seasons may have different reliability

**How Datasets Will Be Combined:**
Dataset #1 (game logs) will be aggregated to team-season level, calculating rest metrics from game-level patterns. This aggregated data will then be merged with Dataset #2 (season statistics) on Team and Season identifiers, creating a unified dataset where each row represents one team in one season with both rest patterns and performance outcomes.

```python
## Dataset 2: Load, Tidy, Analyze, Clean, and Save

# Load team season statistics from Basketball Reference
print("\n" + "="*70)
print("DATASET #2: NBA TEAM SEASON STATISTICS")
print("="*70)

# 4A - Load the data
print("\n4A - LOADING DATA")
print("-" * 70)

url_season = 'https://www.basketball-reference.com/leagues/NBA_2024.html'

try:
    tables_season = pd.read_html(url_season)
    df_season = tables_season[0]  # First table is team stats
    print(f"Successfully loaded team season statistics")
    print(f"Initial shape: {df_season.shape}")
except Exception as e:
    print(f"Error loading data: {e}")
    df_season = pd.DataFrame()

print("\nFirst few rows of raw data:")
print(df_season.head())

# 4B - Tidy the data
print("\n4B - TIDYING DATA")
print("-" * 70)

# Remove any duplicate header rows
df_season_tidy = df_season[df_season.iloc[:, 0] != 'Rk'].copy()
print(f"Removed duplicate header rows")
print(f"Shape after removing headers: {df_season_tidy.shape}")

# Rename the first column to a stable team identifier
first_col = df_season_tidy.columns[0]
df_season_tidy = df_season_tidy.rename(columns={first_col: 'team_name'})
print(f"Renamed first column '{first_col}' to 'team_name'")

# Convert numeric columns to proper types
candidate_numeric_cols = ['W', 'L', 'PW', 'PL', 'Pace', 'ORtg', 'DRtg', 'SRS', 'Age', 'W/L%', 'PS/G', 'PA/G', 'GB']
numeric_cols = [col for col in candidate_numeric_cols if col in df_season_tidy.columns]
for col in numeric_cols:
    df_season_tidy[col] = pd.to_numeric(df_season_tidy[col], errors='coerce')

print("Columns in tidied data:")
print(df_season_tidy.columns.tolist())

# 4C - Dataset size
print("\n4C - DATASET SIZE")
print("-" * 70)
print(f"Number of teams: {df_season_tidy.shape[0]}")
print(f"Number of variables: {df_season_tidy.shape[1]}")
print(f"Total data points: {df_season_tidy.shape[0] * df_season_tidy.shape[1]}")

# 4D - Data quality overview
print("\n4D - DATA QUALITY OVERVIEW")
print("-" * 70)
print("No missingness analysis is displayed for this section.")

# 4E - Outlier detection
print("\n4E - OUTLIER DETECTION")
print("-" * 70)

print("\nSummary statistics for key metrics:")
available_numeric_cols = [col for col in numeric_cols if col in df_season_tidy.columns]
if available_numeric_cols:
    summary_stats = df_season_tidy[available_numeric_cols].describe()
    print(f"Using numeric columns: {available_numeric_cols}")
else:
    summary_stats = pd.DataFrame()
    print("No numeric columns available for summary statistics.")
print(summary_stats)

# Detect outliers using IQR method
print("\n--- Outliers detected (IQR method: 1.5 * IQR) ---")
outlier_count = 0
for col in available_numeric_cols:
    if df_season_tidy[col].dtype in ['float64', 'int64']:
        Q1 = df_season_tidy[col].quantile(0.25)
        Q3 = df_season_tidy[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers_mask = (df_season_tidy[col] < lower_bound) | (df_season_tidy[col] > upper_bound)
        num_outliers = outliers_mask.sum()

        if num_outliers > 0:
            print(f"\n{col}: {num_outliers} outlier(s) outside [{lower_bound:.2f}, {upper_bound:.2f}]")
            print(df_season_tidy[outliers_mask][['team_name', col]])
            outlier_count += num_outliers
        else:
            print(f"{col}: No outliers")

if outlier_count == 0:
    print("\nNo significant outliers detected across numeric columns.")

# 4F - Data cleaning
print("\n4F - DATA CLEANING")
print("-" * 70)

df_clean_final = df_season_tidy.copy()

# Remove rows with missing team names
df_clean_final = df_clean_final.dropna(subset=['team_name'])
print(f"Removed {len(df_season_tidy) - len(df_clean_final)} rows with missing team names")

# Strategy for missing data:
print("\nMissing data handling strategy:")

# Fill Pace and Age with league average (important control variables)
if 'Pace' in df_clean_final.columns:
    pace_mean = df_clean_final['Pace'].mean()
    df_clean_final['Pace'].fillna(pace_mean, inplace=True)
    print(f"  - Imputed missing 'Pace' with league average ({pace_mean:.2f})")

if 'Age' in df_clean_final.columns:
    age_mean = df_clean_final['Age'].mean()
    df_clean_final['Age'].fillna(age_mean, inplace=True)
    print(f"  - Imputed missing 'Age' with league average ({age_mean:.2f})")

# Fill playoff stats with 0 (didn't make playoffs)
for col in ['PW', 'PL']:
    if col in df_clean_final.columns:
        df_clean_final[col].fillna(0, inplace=True)
        print(f"  - Set missing '{col}' to 0 (team didn't make postseason)")

# Drop rows with missing core outcome variables (W/L)
if 'W' in df_clean_final.columns and 'L' in df_clean_final.columns:
    df_clean_final = df_clean_final.dropna(subset=['W', 'L'])
    print(f"  - Removed rows with missing win/loss data (core outcome variable)")

print(f"\nFinal dataset after cleaning: {df_clean_final.shape[0]} teams × {df_clean_final.shape[1]} columns")
print(f"\nMissing values after cleaning:")
remaining_missing = df_clean_final.isnull().sum()
if remaining_missing.sum() > 0:
    print(remaining_missing[remaining_missing > 0])
else:
    print("  No missing values remaining!")

# 4G - Save processed data
print("\n4G - SAVING PROCESSED DATA")
print("-" * 70)

os.makedirs('data/02-processed', exist_ok=True)

df_clean_final.to_csv('data/02-processed/nba_team_stats_2024.csv', index=False)
file_size = os.path.getsize('data/02-processed/nba_team_stats_2024.csv') / 1024

print(f"✓ Saved to: data/02-processed/nba_team_stats_2024.csv")
print(f"  - {df_clean_final.shape[0]} rows (teams)")
print(f"  - {df_clean_final.shape[1]} columns")
print(f"  - File size: {file_size:.1f} KB")

print("\n" + "="*70)
print("DATASETS 1 & 2 INTEGRATION SUMMARY")
print("="*70)
print(f"\nDataset 1 (Game logs):")
print(f"  - Unit: Individual game")
print(f"  - Purpose: Identify game patterns for rest analysis")
print(f"\nDataset 2 (Team season stats):")
print(f"  - Unit: Team-season")
print(f"  - Rows: {df_clean_final.shape[0]} teams")
print(f"  - Purpose: Outcome variables and control variables")
print(f"\nNext step: Aggregate Dataset 1 to team-season level and merge with Dataset 2")
print("="*70)
```
